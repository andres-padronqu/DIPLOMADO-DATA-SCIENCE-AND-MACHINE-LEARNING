     **1.  Defina una red de aprendizaje profundo (Deep Learning)**
   
Es un modelo de machine learning basado en redes neuronales artificiales con muchas capas de procesamiento (capas ocultas). Cada capa transforma las variables de entrada en representaciones cada vez más abstractas mediante combinaciones lineales de los pesos y funciones de activación no lineales. Al apilar varias capas, la red es capaz de modelar relaciones complejas y no lineales en los datos y aprender automáticamente representaciones útiles para tareas como clasificación, regresión o detección de anomalías.

    3.  Describe las funciones de los tres grupos de capas de una red de aprendizaje profundo. ¿Cómo sería la capa de Output? Recuerda que en clasificación cada nodo puede ser representado como una categoría .

Una red profunda clásica tiene tres tipos de capas:

- **Capa de entrada (Input layer):** Recibe las variables originales: métricas de rendimiento de los gestores, exposiciones a factores de estilo, rotación de portafolio, volatilidad, etc.

Solo presenta los datos al modelo, no hace transformación más allá de normalizaciones.

- **Capas ocultas (Hidden layers):** Conjunto de una o varias capas densamente conectadas.

Cada neurona calcula una combinación lineal de las salidas anteriores más una función de activación no lineal (ReLU, tanh, etc.).

Su función es aprender representaciones internas que capturen combinaciones complejas de las características de los gestores: patrones de trading, sensibilidad a factores, firmas de estilo, etc.

- **Capa de salida (Output layer):** Traduce la última representación interna en la predicción final.
Podría ser:

Una capa densa con un nodo por cada categoría de estilo.

Activación softmax, para que cada nodo entregue la probabilidad de que el gestor pertenezca a ese estilo.

La clase predicha sería la del nodo con mayor probabilidad.

De esta forma, cada nodo de salida representa una categoría y el modelo también puede indicar si un gestor está cambiando de estilo.

### Parte 2
Investigar la función Soft-Max. 
1. ¿Hacia qué dominio mapea X?

La función **Softmax** toma un vector de números reales  
$$x = (x_1,\dots,x_K) \in \mathbb{R}^K$$ y lo mapea al **espacio de probabilidades**:

$$
\text{softmax}(x)_i = \frac{e^{x_i}}{\sum_{j=1}^K e^{x_j}}, \quad i = 1, \dots, K.
$$

- **Dominio:** $$\mathbb{R}^K$$ (cualquier vector de logits).  
- **Codominio:** $$\Delta^{K-1} = \{\,p \in \mathbb{R}^K : p_i \ge 0,\ \sum_{i=1}^K p_i = 1\,\},$$  
  es decir, vectores de **probabilidades**.

2. ¿Cómo lo graficarían en Python?
Podemos ilustrar la función Softmax para 3 clases dejando fijos los logits
de dos clases en 0 y variando el logit de la clase 1. Luego graficamos las
probabilidades resultantes de cada clase.

```python
# Importamos librerías

import numpy as np
import matplotlib.pyplot as plt

def softmax(x):
    """
    Calcula softmax columna por columna.
    x: arreglo de shape (K, n_puntos)
    """
    e_x = np.exp(x - np.max(x, axis=0, keepdims=True))  # para mantener estabilidad numérica
    return e_x / e_x.sum(axis=0, keepdims=True)

# Rango de valores para el logit de la clase 1
z = np.linspace(-5, 5, 200)

# 3 clases: la clase 1 varía con z, las otras dos se mantienen en 0
logits = np.vstack([
    z,
    np.zeros_like(z),
    np.zeros_like(z)
])  # shape (3, 200)

# Probabilidades softmax para cada punto
probs = softmax(logits)

# Gráfica
plt.plot(z, probs[0, :], label="Clase 1")
plt.plot(z, probs[1, :], label="Clase 2")
plt.plot(z, probs[2, :], label="Clase 3")
plt.xlabel("Logit de la clase 1")
plt.ylabel("Probabilidad (Softmax)")
plt.title("Función Softmax para 3 clases")
plt.legend()
plt.grid(True)
plt.show()
```

Se ve cómo, al cambiar el logit de la clase 1, cambian las probabilidades de las tres clases.

3. ¿Por que crées que es útil?

Yo creo que es muy útil ya que:
1. Convierte las salidas de la red en **probabilidades** para cada clase (entre 0 y 1).
2. Es suave y derivable, por lo que se puede entrenar con descenso de gradiente y
   **entropía cruzada**.
3. Resalta al ganador: una clase con logit mayor recibe una probabilidad mucho más alta.
