## Relación entre Regresión Múltiple y Redes Neuronales. 

Panel A: Regresión Hipotética usando 4 features/inputs 1 output. Panel B: Red Neuronal  Básica: 

1. Nodos (Círculos) conectados por Links (flechas conectando Nodos)
2. Redes Neuronales tienen 3 tipos de capas: 
    1. Capa de Entrada (Ejemplo con un nodo para cada Feature)
    2. Capas Ocultas donde el aprendizaje ocurre y los inputs son procesados en redes entrenaas
    3. Capa de Salida (Aquí en un único Nodo para la variable Target y)
    
  

Características Adicionales de las Redes
1. Los nodos en las capas ocultas de ANN transformlos los inputs en una forma no-lineal y luego esos valores son combinados en valor target (Y)
2. En el ejemplo se usa Unidad Lineal Rectificada (Rectfied Lineal Unti: ReLu) f(x)=max(0,x)
3. y=b1(z1)+b2(z2)+b3(z3) y cada z es una ReLu: z1=max(b4x1+b5x2+b6x3,0); zi
4. ANN, los features se escalan para controlar diferencias en las unidades. e.g si los datos fuera positivos se podrían escalar con el máximo para que vayan de 0 a 1. 

## Red Neuronal un poco más compleja

Esta red : 
1. Capa de Entrada con 4 Nodos
2. Una capa oculta con 5 nodos ocutlos
3. Un Nodo de Salida. 

Estos tres son hiperparámetros que determinan la estrucrura de la red. (4-5-1) Red Neuronal

Los Nodos a la derecha del Input, la Capa de los Ocultos, se les **Neuronas** porque procesan la información recibida. 

Por ejemplo, el primer nodo recibe 4 links de los inputos, por lo que le *transmite* a esa neurona 4 valores. 

Cada nodo tiene, concepctualemente, 2 partes funcionales: 

1. **Un operador de Suma**. Recibe los 4 valores, El operador de Suma multiplica da valor por un peso y suma los valores ponderados para un formar el Input Total de Esa Red. El input total passa la función de Activación

2. **Función de Activación** : Transforma este input en el Output final del Nodo. Informalmente, funciona como un switch de brillo que incrementa o disminuye la potencia del input. Las funciones de activación se caracterizan por ser no lineales: 
    1. Sigmoidal (rando del ouput: 0-1)
    2. ReLu
    3. Hiperbólica, entre otras. 
    
El ser no lineal implica que la tasa de cambio de la salida difiere para diferntes nivles del input. 

# Funciones de Activación 

Por ejemplo, la función Sigmoidal (Forma S) convierte una cantidad negativa en la red input en un out cercano a 0. Este Output cercano a 0 implica que no se "dispara", por lo que no hay que pasarle nada a la siguiente neurona. 

Una cantidad positva en la red de input es transformado en un número cercano a 1, por lo que el nodo "dispara". 

El output de la función de activación es transmitido al siguiente conjunto de nodos si es que hay una segunda capa oculpa, o como en la imagen de arriba a la capa de salida. 

Este proceso de transmisión (flechas apuntando hacia adelante) se le conoce como propagación hacia adelante **forward propagation**

# Algoritmo a Grandes Rasgos

Puntos a considerar al Entrenar una Red Neuronal: 
1. Se inicia con un conjunto de pesos aleatorios en la redes 
2. En un Contexto de Aprendizaje Supervisado, entrenar una Red Neuronal es un proceso iterativo en donde las predicciones se comparan con los valores reales y se evalúan con una métrica de perfomance: Error cuadrático Medio, por ejemplo. 
3. Los pesos se van ajustando para reducir el error de la red
4. El aprendizje ocurre en el proceso de ajuste de los pesos de las redes con el objetivo de disminiuir el error
5. Nuevo peso= Viejo Peso - [Tasa de Aprendizaje x Derivada Parcial del Error con respecto al peso viejo]

    5.1 Deriada Parcial es un Gradiente o Tasa de Cambio del Error Total con respecto cambios en el Viejo Peso
    
    5.2 Tasa de Aprendizaje: Parámetro que afecta la magnitud del los ajustes
6. Cuando el aprendijza está completado, todos los pesos en la red son asignados

# Deep Learning

Redes Neuronales con muchas capas ocultas (al menos 3 y muy frecuente más de 20 capas ocultas) se conocen como Redes de Aprendizaje Profundo (**Deep Learning Nets (DLNs)**) y la columna vertebral de la revolución en inteligencia artificial. 

1. Reconocimiento de Imágenes
2. Detección de Fraudes de CArteras de Crédito
3. Visión y Control de Coches Autónomos
4. Procesamiento de Lenguaje Natural (e.g. traducciones)
5. Patrones
6. Reconomiento comunicionaciones (Speech). 


### Algoritmo 
1. Toman un cojunto de inputx x de un grupo de features( inputl layer)
2. Son pasadas a una capa de función matemáticas no lineales (neuronas ) con pesos y cada una de ellas produce un numero escalonada en el rango (0,1) o (-1,1)
3. Estos números son pasados a otra capa con funciones y luego a otra y así hasta que la capa fnal produce produce un conjunto de probabilidades (si fuera supervisado de clasificacioón) que la neuva observación pertenece a alguna cateorgía (cada una representada por un output final). 
4. DLN asigana la categoriían con mayor proba
5. DLN se entrena con grandes data sets, y los pesos se escogen para minimizar (la función de pérdida)

Aunque los nodos en los inputs y las capas de salida son determinadas por las características de los features y de los ouputs en la predicción, se tienen que esocgeer muchos hiperparámetros. 
1. Números de Capas y de Nodos ocultos
2. La Arquitectura de la forma de su conexión y activación

Objetivo: Mejoara el *Out-of-Sample* desempeño, lo cual es una tarea no fácil. Utilizar un Guess razonables en experiencia y después ir calibrando. DLN requieren tiempo para entrenar y variar sistemáticamente sus hiperparámetros no es fácil. Para bases pequeñas uno puede empezar con 2 o tres capas ocultas y cientos de nodos antes de ajustar los parámetros hasta que se tenga un modelo con buen poder predictivo

DLN famosas por: 
1. Avances en métodos analíticos para ajustar estos modelos. 
2. Disponibilidad de de Bases para Entrenar
3. Computadoras más Rapidas (nuevos chips del tipo Unidad de Procesamiento Gráfico (GPU) hechos a la medida pra DLN). 

Aplicaciones en Finanzas: 
1. Trading
2. Automatizar Procesos
3. Valuar Opciones. Utilizando el modelo de Black-Scholes con 6 inputs (spot, strike, time, dividendos,rf, vol) y 4 capas oculatas con 100 neuronas cada una, y un output. Los procios de predición (out-of-sample) tuvieron una r2 de 99.8%
4. DLN con multicapas perceptrones (feed forward networks) con miles de nodos en varias capas ocultas se utlizan paara predecir factores corporativos fundamentales y factores técnicos relacionados al precios. 
5. Utilizan docenas de input Features de Estados financieros e Indicadores de Momentum de Acciones de diferentes periodso (eg. 1, 3 o 6 meses) para generar ideas de trading
6. Estrategias Inversión usando DLN para predecir factores (book value to market value, operating income to market cap,e tc) han alcanzado rendimientos compuestos anuales de 270 puntos base que usando modelos de factores estándar. 

# Reinforcement Learning

* **Agente** que realiza **acciones** que maximizarán sus **beneficios** a través del timpo, tomando en consdieranción su **ambiente**

En el jemplo de AlphaGo: 
1. El jugador virtual (*agente*)
2. Usa los comandos de la consola (*acciones*)

A diferencia de aprendizaje supervisado, RL no tiene directamente etiquetas para cada observación en su ambiente ni retroalimentación instantánea. 

RL necesitas observar su ambiente, aprende probando neuvas acciones (algunas no son óptimas de inmediato) y reutiliza sus experiencias previas. 

El aprendizaje ocurre a través de millones de intentos y errores. 

**En Finanzas** se están utilizando para **estrategias de inversión** donde: 
1. El trader virtual (*agente*)
2. Siguen un conjunto de reglas (*acciones*)
3. En un mercado específico (*ambiente*)
4. Para maximizar sus rendimientos (*beneficios*)

### Una red neuronal de una capa
Una red neuronal de una capa está dada por: 
$$
    \hat y = {\bf w}^{(2)} f^{(1)}({\bf w}^{(1)}x)
$$

```python
#y es la variable respuesta
#x son los features
#f es una función de activación
#w1 y w2 son un conjunto de pesos/flechas/coeficientes
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.random import randn, seed

import sys
sys.path.append("../files")
import nn_utils
```

# Funciones de Activación 
Existen diferentes funciones *comunes* de activación
## Sigmoidal

$$
    \sigma(x) = \frac{1}{1 + \exp(-x)}
$$

```python
x=np.linspace(-5,5,500)
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
plt.plot(x,sigmoid(x))
```

# Tangente Hipérbólica
$$
    \tanh(x) = \frac{\exp(x) - \exp(-x)}{\exp(x) + \exp(-x)}
$$

```python
def tanh(x):
    return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))
plt.plot(x,tanh(x))
```

# ReLU (Rectifiend Linear Unit)
$$
    \text{ReLU}(x) = \max(0, x)
$$

```python
def ReLu(x):
    return np.maximum(0,x)
plt.plot(x,ReLu(x))
```

# Definición de Red Neuronal
$$
    \hat y = {\bf w}^{(2)} f^{(1)}({\bf w}^{(1)}x)
$$

```python
def nnet(x,W1,W2): return W2@sigmoid(W1@x)
#y es la variable respuesta
#x son los features
#f es una función de activación
#w1 y w2 son un conjunto de pesos/flechas/coeficientes
```

```python
#Datos 
#4 características/columnas/features x0,x1,x2,x3
#1,000 observaciones
seed(31415)
nobs=1000
x=randn(4,nobs) #hay 4 inputs (variabls explicativas,regresres/features)
y=randn(1,nobs)
```

```python
#Inicializar los pesos
#vamos a proponer 3 neuronas en la única capa oculta
w1=randn(3,4) #aquí hay 12 pesos/flechas (conectan cada una de las 3 neuronas
#en la única capa oculta con los 4 inputs)

w2=randn(1,3) #aquí hay 3 flechas (conectan el único output con 
#con las 3 neuronas ocultas)
```

```python
w1 #numero de fila es la neurona de la capa oculta
#el número de columa es el input que se conecta. 
```

```python
x.shape
```

```python
(w1@x).shape #3 neuronas
#dentro de cada neurona está el caso de las 100 observaciones
#para una observación/fila particular tendrían 1x3 #3neuronas, a cada
#neuornas le sacaríamos la función de activación
#ese resultado lo multiplicamos por unvector w2 de pesos (3x1)
#el producto les va a generar algo de 1x1 (real)

#si lo hacen para 1000 observaciones les va a generar 1x1000 (un vector 
#de 1000 pronósticos de su red neuronal.)
```

```python
#vamos a hace 4 gráficas. Estmaos tomando 8 flechas al azar de 
#nuestra red para ver la interacción

cuts=[((2,0),(1,0)),
     ((0,1),(0,3)),
     ((1,2),(2,3)),
     ((2,3),(2,2))]
```

```python
fig=plt.figure(figsize=(10,7.5))
for ix, cut in enumerate(cuts):
    ax=fig.add_subplot(2,2,ix+1)
    p1,p2,J=nn_utils.loss_surface((x,w1,w2),(w1,w1),cut,nnet,y)
    #primer argumento: son los parámetros de la red neuronal
    #segundo argumento: son los pesos/flechas que se van a comparar
    #en este caso queremos los pesos que conectan el input con la primer 
    #capa oculta w1
    #tercer argumento: son las coordenadas que vamos a evaluar
    #cuarto modelo: modelo de la red neuronal
    #quinto argumento: y_true, para sacar los errores/costo asociados que 
    #al final es lo que estamos graficando
    
    plt.contourf(p1,p2,J,40,cmap="bone_r")
    
    #información a la gráfica
    
    (cut1_ix,cut1_layer),(cut2_ix,cut2_layer)=cut
    
    
    #super indice : haga referencia al input
    #sub indice: haga referencia a la neurona de la capa oculta
    
    xlab="$W^{[" + f"{cut1_layer+1}" + "]}"  + f"_{cut1_ix+1}$"

    plt.xlabel(xlab,fontsize=30)
    
    ylab="$W^{[" + f"{cut2_layer+1}" + "]}"  + f"_{cut2_ix+1}$"

    plt.ylabel(ylab,fontsize=30)
    
    plt.colorbar()
    
    
    
    
plt.tight_layout()  
    
    
    
```
