### Regresion lineal multiple

En la sesión anterior introdujimos el modelo de regresión lineal simple. Con el que estimamos la altura de los hijos con la altura de los padre como variable independiente. 
En este caso nos limitamos a predecir nuestra variable objetivo con una unica variable predictora.

Qué es lo que pasa cuando queremos incluir más de una variable para predecir una variable objetivo?

En este caso debemos ajustar un modelo de Regresión Lineal Multiple

En este caso, el problema sería encontrar $\theta_0, \theta_1, \ldots, \theta_m$. Nuestra predicción para la $i$-ésima observación $\hat y$ estaría dada por
$$
    \begin{align}
    \hat y_i &= \theta_0 + \theta_1 x_{1} +  \theta_2 x_2 + \ldots + \theta_m x_m\\m
            &= \theta^T x^i
    \end{align}
$$

Donde
$$
    \theta = \begin{bmatrix}
                \theta_0 \\
                \theta_1 \\
                \theta_2 \\
                \vdots  \\
                \theta_m
             \end{bmatrix}
$$

$$
    x_i = \begin{bmatrix}
                1 \\
                x_{1,i} \\
                x_{2,i} \\
                \vdots  \\
                x_{m,i}
             \end{bmatrix}
$$

Definimos nuestro *training* dataset como la siguiente representación matricial:
$$
    X = \begin{bmatrix}
        – x_1 – \\
        – x_2 – \\
        \vdots \\
        – x_n – \\
        \end{bmatrix}
        = \begin{bmatrix}
                1 & x_{1,1} & x_{1,2} & \ldots & x_{1,m}\\
                1 & x_{2,1} & x_{2,2} & \ldots & x_{2,m}\\
                \vdots & \vdots & \vdots & \ddots & \vdots\\
                1 & x_{n,1} & x_{n,2} & \ldots & x_{n,m}\\
          \end{bmatrix}
$$

siendo los targets:
$$
    y = \begin{bmatrix} y_1 & y_2 & \ldots & y_n \end{bmatrix}
$$

Para un problema de **regresión lineal multiple (Generalizada)**, el vector $\theta^* = [\theta_0, \theta_1, \ldots, \theta_m]$ que minimiza el problema de minimos cuadrados son:

$$
    \theta^* = (X^T X)^{-1} X^T y
$$

Veamos un ejemplo de como ajustar un modelo de regresión multiple con un conjunto de datos reales del Inegi, donde El objetivo es modelar el ingreso  promedio en los hogares de los municipios del país a partir un conjunto de variables sociodemograficas relevantes.

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.3, random_state=42)
```

Para medir el desempeño del modelo utilizaremos el coeficiente de determinación que se expresa como sigue:
$$
    \mbox{R}^2  = 1 -  \frac{\sum_i^n(y_i - \hat y_i)^2}{\sum_i^n(y_i - \bar y)^2}
$$

Esta medida nos dice que tan cerca estan los valores reales del modelo de regresión. También podemos interpretarlo como "que tanta variabilidad de los datos logra explicar nuestro modelo" toma valores de 0 a 1 y buscamos que r2 sea cercano a uno

```python
lin_reg =  LinearRegression().fit(X_train,Y_train)
y_pred = lin_reg.predict(X_test)
y_pred_train =  lin_reg.predict(X_train)
print(np.round(lin_reg.score(X_train, Y_train),5))
print(np.round(lin_reg.score(X_test, Y_test),5))
```

Del resultado anterior vemos como nuestro modelo de regresión lineal explica casi un 60% de variabilidad de los datos de pruena y solo 14% en el conjunto de prueba. Recordemos que tener una diferencia grande entre el desempeño en el conjunto de entrenamiento y de prueba es un indicador de overfitting

veamos el resultado de los estimadores

```python
coefs_linreg=pd.DataFrame([X_train.columns.tolist(),lin_reg.coef_]).T
coefs_linreg.columns=["variable","coeficiente_lingreg"]
coefs_linreg 
```

### Regularización

#### - Motivacion

Recordemos que cuando ajustamos un modelo, este puede tener overfitting o underfitting. Esto se puede ver de la siguiente manera:
    


<img src="OVERFITTING_underfitting.png">

En ocasiones, el modelo con el cual trabajamos es muy complejo si la base de datos con la cuál estamos trabajando cuenta con un gran número de columnas .Al reducir la complejidad del modelo lograremos reducir el overfitting. 

¿Cómo podemos pasar de un modelo como el que tenemos al centro de la anterior imagen a un modelo como el que tenemos en la tercer imagen? 

En este caso utilizaremos las tecnicas de regularización para tener un modelo más simple y con mejor generalización. En otras palabras, buscamos hacer los coeficientes  $\theta_3, \theta_4 $iguales a cero

 Para una regresión lineal, existen dos maneras de regularizar el modelo: 

**Lasso** (L1) `from sklearn.linear_model import Lasso` minimizamos la siguiente función de perdida


<img src="lasso.png">

**Ridge**  (L2) `from sklearn.linear_model import Ridge` minimizamos la siguiente función de perdida

<img src="Ridge.png">

Veamos que las funciones son el mse que ya conocemos, pero agregamos un ultimo termino (color rojo) este termino penalizará los pesos de los coeficientes estimados. Es nuestra tarea encontrar el mejor valor de la variable $ 0 < \lambda  < inf $ y esta determinará que tanto queremos penalizar los coeficientes. un $ \lambda $ grande penalizará mucho mientras que uno pequeño penalizará poco. Geometricamente , nos encargamos de hacer las pendientes $ \theta_i $ cada vez más pequeñas o empinadas.

La diferencia entre lasso y ridge. Es que el primero al tomar el valor absoluto, puede llevar los pesos a cero, mientras que ridge al tomar el valor al cuadrado acerca los pesos de los coeficientes a cero, pero nunca a cero. De esta forma Lasso puede ser util para seleccionar variables significativas en el modelo


La elección entre L1 y L2 dependerá de la naturaleza del problema y del conocimiento a priori que se tenga sobre los datos. Si se sospecha que solo un subconjunto pequeño de variables es relevante, Lasso puede ser preferido. Si se espera que muchas variables contribuyan de manera moderada, Ridge puede ser más apropiado.

### Elastic Net


Elastic Net es un método de regularización utilizado en problemas de regresión que combina las penalizaciones L1 (Lasso) y L2 (Ridge). Fue propuesto para abordar algunas limitaciones individuales de Lasso y Ridge y aprovechar las ventajas de ambos.
Elastic Net es útil en situaciones en las que se sospecha que hay múltiples variables relevantes y correlacionadas, y se busca un modelo más robusto que combina las ventajas de Lasso y Ridge.

$$
MSE = \frac{1}{n} \sum_{i=1}^{N} (y_i - X_i\beta)^2 + \alpha (\lambda \sum_{j=1}^{p} |\beta_j| + (1-\lambda)\sum_{j=1}^{p} \beta_j^2)
$$

```python
# l1= .3 y l2=  .7
lin_reg_elastic =  ElasticNet(alpha=.002,l1_ratio=.3).fit(X_train,Y_train)
y_pred = lin_reg_elastic.predict(X_test)
y_pred_train =  lin_reg_elastic.predict(X_train)
print("r2 en conjunto de train ",np.round(lin_reg_elastic.score(X_train, Y_train),5))
print("r2 en conjunto de train ",np.round(lin_reg_elastic.score(X_test, Y_test),5))
```

```python
coefs_linreg_elastic=pd.DataFrame([X_train.columns.tolist(),lin_reg_elastic.coef_]).T
coefs_linreg_elastic.columns=["variable","coeficiente_elastic"]
coefs_linreg_elastic
```

podemos ver que los coeficientes de algunas variables se fueron a cero, pero no todas como las que se penalizaron en lasso

veamos graficamente el ajuste de los coeficientes en los 3 modelos

```python
coefs=pd.concat([coefs_linreg,coefs_linreg_lasso,coefs_linreg_ridge,coefs_linreg_elastic],axis=1)
coefs = coefs.loc[:,~coefs.columns.duplicated()].copy()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(coefs['variable'], coefs['coeficiente_lingreg'], label='Linear Regression', marker='o')
plt.scatter(coefs['variable'], coefs['coeficiente_lasso'], label='Lasso', marker='o')
plt.scatter(coefs['variable'], coefs['coeficiente_ridge'], label='Ridge', marker='o')
plt.scatter(coefs['variable'], coefs['coeficiente_elastic'], label='Elastic Net', marker='o')

# Configurar leyendas y etiquetas
plt.legend()
plt.title('Coeficientes de Regresión para Diferentes Métodos de Regularización')
plt.xlabel('Variable')
plt.ylabel('Coeficiente')
plt.xticks(rotation=45)

# Mostrar el gráfico
plt.tight_layout()
plt.show()
```

### Hiperparametrización de $ \lambda $ en Ridge y Lasso

Un hiperparámetro es un componente del modelo cuyo valor se establece antes de entrenarlo y no se ajusta durante el proceso de entrenamiento.los hiperparámetros son configuraciones externas al modelo que influyen en el rendimiento y comportamiento del modelo durante el entrenamiento. Estos hiperparámetros no son aprendidos directamente del conjunto de datos, sino que deben ser establecidos de antemano por el practicante o mediante técnicas de búsqueda. En los metodos de Ridge y Lasso  $ \lambda $  es un hiperparametro. A continuación nos enfocaremos en encontrar el mejor valor en cada caso 

```python
r2_train_lasso=[]
r2_test_lasso=[]
alpha_lasso=[]
for i in np.arange(0.00001,.2,.01):
    
    lin_reg =  Lasso(alpha=i).fit(X_train,Y_train)
    r2_train_lasso.append(lin_reg.score(X_train, Y_train))
    r2_test_lasso.append(lin_reg.score(X_test, Y_test))
    alpha_lasso.append(i)
    
results_lasso=pd.DataFrame()
results_lasso["alpha"]=alpha_lasso
results_lasso["r2_train"]=r2_train_lasso
results_lasso["r2_test"]=r2_test_lasso
#results_lasso["dif_r2"]= results_lasso["r2_train"] - results_lasso["r2_test"]
    
display(results_lasso)#.sort_values(by="dif_r2",ascending=False)


plt.plot(alpha_lasso,r2_train_lasso,label="train")#lineas
plt.scatter(alpha_lasso,r2_train_lasso)#puntos

plt.plot(alpha_lasso,r2_test_lasso,label="test")#lineas
plt.scatter(alpha_lasso,r2_test_lasso)#puntos


plt.xlabel("r2")
plt.ylabel("alpha")

plt.legend()
plt.grid()
```

```python
r2_train_ridge=[]
r2_test_ridge=[]
alpha_ridge=[]
for i in np.arange(0.1,15,.1):
    
    lin_reg =  Ridge(alpha=i).fit(X_train,Y_train)
    r2_train_ridge.append(lin_reg.score(X_train, Y_train))
    r2_test_ridge.append(lin_reg.score(X_test, Y_test))
    alpha_ridge.append(i)
    
results_ridge=pd.DataFrame()
results_ridge["alpha"]=alpha_ridge
results_ridge["r2_train"]=r2_train_ridge
results_ridge["r2_test"]=r2_test_ridge
#results_ridge["dif_r2"]= results_ridge["r2_train"] - results_ridge["r2_test"]
    
#results_ridge.sort_values(by="dif_r2",ascending=False)
display(results_ridge)





plt.plot(alpha_ridge,r2_train_ridge,label="train")#lineas
plt.scatter(alpha_ridge,r2_train_ridge)#puntos

plt.plot(alpha_ridge,r2_test_ridge,label="test")#lineas
plt.scatter(alpha_ridge,r2_test_ridge)#puntos


plt.xlabel("r2")
plt.ylabel("alpha")

plt.legend()
plt.grid()
```
