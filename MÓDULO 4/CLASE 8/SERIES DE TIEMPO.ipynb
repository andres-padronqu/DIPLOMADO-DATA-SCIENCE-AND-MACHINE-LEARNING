Las series de tiempo son conjuntos de datos ordenados temporalmente, donde las observaciones se registran en intervalos regulares. Desde un punto de vista probabilistico , una serie de tiempo es la realización de un proceso estocástico.
La modelización de series temporales implica prever valores futuros basándose en patrones pasados. A diferencia de los problemas de clasificación y regresión que hemos ajustado anteriormente, en el analisis de series de tiempo nos encargaremos de hacer predicciones con base en una sola variable: El tiempo ($t$). Visto así queremos estimar un valor en función del tiempo $y=X(t)$

Una serie de tiempo se puede dividir en los siguientes componentes:

**Componente de tendencia**
Una tendencia es un movimiento direccional constante en una serie de tiempo. Estas tendencias serán deterministas o estocásticas. El primero nos permite proporcionar un fundamento para la tendencia, mientras que el segundo es una característica aleatoria de una serie que es poco probable que expliquemos. Las tendencias suelen aparecer en series financieras y muchos modelos comerciales utilizan sofisticados algoritmos de identificación de tendencias.

<img src="serie tendencia.png">


**Componente estacional**
Muchas series de tiempo contienen variaciones estacionales. Esto es particularmente cierto en las series que representan las ventas comerciales o los niveles climáticos. En las finanzas cuantitativas, a menudo vemos variaciones estacionales, particularmente en series relacionadas con las temporadas de vacaciones o la variación anual de la temperatura (como el gas natural).

Podemos escribir las componentes de una serie de tiempo yt como 

$yt = St + Tt + Rt$

donde $St$ es el componente estacional, $Tt$ es el componente de tendencia y $Rt$ representa el componente restante de la serie de tiempo no capturado por el componente de tendencia o estacional.

## Definiciones clave en Series de tiempo

#### Ruido blanco

El ruido blanco es un concepto fundamental en estadísticas y series temporales. Se refiere a un tipo específico de secuencia de datos estocásticos que exhibe propiedades particulares. Aquí hay algunas características clave del ruido blanco:

- Media Constante:

La media del ruido blanco es constante y generalmente se establece en cero.

- Varianza Constante:

La varianza del ruido blanco es constante para todos los puntos de datos.

- Independencia:

Las observaciones de ruido blanco son independientes entre sí. La presencia o ausencia de un valor no proporciona información sobre los valores futuros o pasados.

-  Distribución Normal:

Las observaciones de ruido blanco se distribuyen de manera normal. Sin embargo, esta característica no es siempre estrictamente necesaria; en muchos casos, se asume que las observaciones son aproximadamente normales.


En resumen, el ruido blanco es un proceso estocástico que representa la aleatoriedad pura sin patrones o estructuras discernibles. A menudo, se utiliza como un modelo de base para describir el componente de error o residuo en muchos modelos estadísticos y de series temporales. Cuando los residuos de un modelo se comportan como ruido blanco, es un indicativo de que el modelo está capturando adecuadamente la variabilidad aleatoria en los datos.

<img src="ruido blanco.png">

##### Serie estacionaria

Una serie temporal se considera estacionaria cuando sus propiedades estadísticas no cambian con el tiempo. Si  la serie no tiene cambios sistemáticos en la media (no hay tendencia de la serie), si no tiene cambios sistematicos en la varianza (la varianza es constante a lo
largo del tiempo) y si no hay variaciones periodicas (no tiene ciclos) En otras palabras, la estacionariedad implica que la serie mantiene una estructura constante a lo largo de diferentes intervalos de tiempo. Así, una serie de tiempo con tendencia o con estacionalidad no es estacionaria, ya que la tendencia y la estacionalidad afectarán el valor de la serie de tiempo en diferentes momentos. Por otro lado, una serie de ruido blanco es estacionaria, ya que no importa cuando la observe; debería verse similar en cualquier momento.

Para muchos métodos de modelado de series temporales, como los modelos ARIMA (Autorregresivo Integrado de Medias Móviles), la estacionariedad es un requisito. Los modelos asumen que la serie es estacionaria o se puede convertir en estacionaria 




<img src="serie estacionaria.png">

#### prueba de Dickey-Fuller 

es una prueba estadística utilizada para evaluar si una serie temporal no es estacionaria, lo que puede afectar la capacidad de modelar y predecir la serie de manera precisa 


$Ho:$ La serie no es estacionaria

$H1:$ La serie es estacionaria

El resultado de la prueba se compara con un valor crítico y se toma una decisión sobre si rechazar o no la hipótesis nula. Si el valor p (p-value) es lo suficientemente pequeño (generalmente menor que un umbral específico, como 0.05), se puede rechazar la hipótesis nula, lo que sugiere que la serie temporal es estacionaria.

#### Diferenciacion


La diferenciación es uno de los métodos utilizados para hacer estacionaria una serie de tiempo. En este método, calculamos la diferencia de términos consecutivos en la serie. 

La diferenciación se realiza típicamente para deshacerse de la media variable. Matemáticamente, la diferenciación se puede escribir como:

$y_t ′ = y_t - y_{\substack{t -1}}$

donde yt es el valor en un tiempo t.

Además de la diferenciación, otra de las alternativas para hacer la serie estacionaria es aplicar la función logaritmo
a la serie o alguna de las transformaciones de la familia box cox


### Autocorrelación
Hay muchas situaciones en las que los elementos consecutivos de una serie de tiempo muestran una correlación. Es decir, el comportamiento de los puntos secuenciales de la serie se afecta entre sí de manera dependiente. La autocorrelación es la similitud entre observaciones en función del desfase temporal entre ellas. 

# MODELOS TRADICIONALES DE SERIES DE TIEMPO (AR) (MA) (ARIMA)

#### MODELO AUTOREGRESIVO AR(p)

El modelo autoregresivo (AR) es un tipo de modelo de series temporales que describe una variable en función de sus valores pasados, es decir, se basa en su propia historia. En un modelo AR(p), la variable en el tiempo actual depende de sus 
p valores anteriores, más un término de error.

- Ventajas:

Captura la Dependencia Temporal: Los modelos AR son capaces de capturar patrones temporales y dependencias en los datos a través de la historia de la serie temporal.

Simplicidad Conceptual: Son relativamente sencillos de entender y explicar, ya que se basan en la idea de que el valor actual depende de sus valores pasados.

- Desventajas:

Sensibilidad al Orden p: La elección del orden p puede ser crítica. Un orden demasiado bajo podría no capturar patrones temporales importantes, mientras que un orden demasiado alto podría llevar a la sobreajustar el modelo.

Requiere Estacionariedad: Idealmente, el proceso subyacente debe ser estacionario para que el modelo AR sea apropiado.

# modelo de medias móviles MA (q) 

es un tipo de modelo de series temporales que modela una variable en función de los términos de error pasados. En un modelo MA(q), la variable en el tiempo actual depende de los q términos de error anteriores, más un término de error actual. 

Dichos errores son los errores del modelo AR

<img src="ERRORES_MA.png">


En este modelo el error $\epsilon_t$ se puede representar como ruido blanco


- Ventajas:
Flexibilidad para Modelar Tendencias: Los modelos MA son útiles para modelar componentes de error que pueden representar cambios no capturados por las tendencias o patrones en los datos.

Menos Sensibles a la Estacionariedad: A diferencia de los modelos AR, los modelos MA pueden manejar series temporales no estacionarias sin necesidad de diferenciación.

- Desventajas:
No Captura Tendencias Temporales: Los modelos MA son más efectivos para modelar la variabilidad no sistemática en los datos y pueden no capturar bien patrones temporales complejos.

Sensibilidad al Orden q: La elección del orden q es crítica, y un valor demasiado alto podría llevar a la sobreajustar el modelo.

Requiere Estacionariedad: Idealmente, el proceso subyacente debe ser estacionario para que el modelo MA sea apropiado.

## Modelo autorregresivo integrado de promedio móvil ARIMA(p,d,q)

Si combinamos la estacionariedad con la autorregresión y un modelo de media móvil, obtenemos un modelo ARIMA. ARIMA es un acrónimo de AutoRegressive Integrated Moving Average, y tiene los siguientes componentes:

- AR (p) Representa la autorregresión
- MA (q) Representa media móvil
- I(d) Representa el orden de integración. Es simplemente el número de diferencias necesarias para que la serie sea estacionaria.

<img src="MODARIMA.png">


de esta forma un modelo $AR(1)$ = $ARIMA(1,0,0)$

de esta forma un modelo $MA(3)$ = $ARIMA(0,0,3)$

de esta forma un modelo $MA(2)$ con un grado de diferenciación para hacer la serie estacionaria es = $ARIMA(0,1,2)$


Para todos los modelos de la familia Arima necesitamos tener una muestra suficientemente grande, la literatura sugiere un $n>80$

### AR(2) MANUAL

```python

# Generar una serie temporal AR(2) simulada
np.random.seed(42)
n_samples = 100
epsilon = np.random.normal(size=n_samples)
y = np.zeros_like(epsilon)

# Coeficientes AR
phi_1, phi_2 = 0.7, -0.3

# Construir la serie temporal AR(2)
for t in range(2, n_samples):
    y[t] = phi_1 * y[t-1] + phi_2 * y[t-2] + epsilon[t]

# Visualizar la serie temporal generada y la serie original
plt.plot(y, label='Serie Temporal Generada (AR(2))', linestyle='dashed')
plt.xlabel('Tiempo')
plt.ylabel('Valor')
plt.legend()
plt.show()
```

```python
# Generar una serie temporal MA(2) simulada
np.random.seed(40)
n_samples = 100
epsilon = np.random.normal(size=n_samples)
y_ma = np.zeros_like(epsilon)

# Coeficientes MA
theta_1, theta_2 = 0.5, -0.2

# Construir la serie temporal MA(2)
for t in range(2, n_samples):
    y_ma[t] = epsilon[t] + theta_1 * epsilon[t-1] + theta_2 * epsilon[t-2]

# Visualizar la serie temporal generada MA(2)
plt.plot(y_ma, label='Serie Temporal Generada (MA(2))', linestyle='dashed',color="red")

plt.xlabel('Tiempo')
plt.ylabel('Valor')
plt.legend()
plt.show()
```

# GRAFICOS DE AUTOCORRELACION (ACF) Y AUTOCORRELACION PARCIAL (PACF)

El ACF mide la correlación entre una serie temporal y sus valores rezagados (anteriores).

El PACF mide la correlación entre una serie temporal y sus valores rezagados, controlando la influencia de los rezagos intermedios.

En un gráfico de ACF, el eje horizontal representa el rezago (lag), y el eje vertical muestra el valor de la correlación. Cada barra en el gráfico indica la correlación entre la serie y sus rezagos en un punto específico en el tiempo.

En un gráfico de PACF, el eje horizontal representa el rezago, y el eje vertical muestra el valor de la correlación parcial. Cada barra indica la correlación parcial entre la serie y su rezago correspondiente, eliminando la influencia de los rezagos intermedios.

En otras palabras El ACF traza la correlación entre los datos y sus valores retrasados, mientras que el PACF traza la correlación entre los datos y sus valores retrasados después de eliminar el efecto de los valores intermedios.

La autocorrelación (ACF) y la autocorrelación parcial (PACF) son medidas de asociación entre valores de serie actuales y pasadas e indican cuáles son los valores de series pasadas más útiles para predecir valores futuros. 


Para encontrar el orden de un modelo MA y AR  se puede buscar patrones en los gráficos ACF y PACF, como picos, cortes o colas, que indiquen los posibles valores de p y q. Por ejemplo, si el ACF se corta después del retraso k, entonces q en un modelo MA podría ser k o inferior. Si el PACF se corta después del retraso k, entonces p en un Modelo AR podría ser k o inferior.

#### PARENTESIS (TEST DE CAUSALIDAD DE GRANGER)

El test de causalidad de Granger es una prueba estadística que evalúa si una serie temporal puede predecir otra serie temporal. La prueba se utiliza para examinar si las observaciones pasadas de una variable son útiles para predecir otra variable en el futuro. 

$Ho:$ La serie temporal X no Granger-causa la serie temporal Y. En otras palabras, la información pasada de X no proporciona información adicional para predecir Y


$H1:$ La serie temporal X Granger-causa la serie temporal Y. La información pasada de X es útil para predecir Y, más allá de la información proporcionada por la historia pasada de Y.

$maxlag$: Especifica el número máximo de retardos que se incluirán en la prueba. Un retraso representa un paso en el tiempo. Al establecer max_lag, decides cuántos pasos en el tiempo hacia atrás se considerarán al evaluar la relación de causalidad.


```python
grangercausalitytests(df1[['AveragePrice','Total Volume']],maxlag=[12]) ## dado que el p value es menor que .05
### decimos que con base en volumen podemos estimar el precio
```

una forma de evitar la estabilización del modelo es utilizar el método de pronóstico continuo, es decir, predecimos un valor único y usamos este valor predicho nuevamente para ajustar el modelo y predecir el siguiente valor.

```python
size = int(len(df1) - 34)


history = [x for x in train]
predictions = []

for t in range(len(test)):
    
    model = sm.tsa.arima.ARIMA(history, order=(2,1,2))
    model_fit = model.fit()
    
    output = model_fit.forecast()
    yhat = output[0]
    predictions.append(float(yhat))
    
    obs = test[t]
    history.append(obs)
    
    #print('predicted = %f, expected = %f' % (np.exp(yhat), np.exp(obs)))
```

# SARIMA (p,d,q)(P,D,Q,s)


El modelo SARIMA (Seasonal AutoRegressive Integrated Moving Average) es una extensión del modelo ARIMA (AutoRegressive Integrated Moving Average) que incluye componentes estacionales. El modelo SARIMA es útil para modelar y predecir series temporales que exhiben patrones estacionales o ciclos recurrentes a lo largo del tiempo.


$P$ (Seasonal AutoRegressive Order): Representa el orden de la parte autorregresiva estacional y especifica cuántos periodos anteriores de la serie deben considerarse en la parte autorregresiva estacional.

$D$ (Seasonal Difference Order): Es similar al componente de diferenciación, pero aplicado a la serie temporal estacional. Indica cuántas veces se debe diferenciar la serie temporal estacional para hacerla estacionaria.

$Q$ (Seasonal Moving Average Order): Representa el orden de la parte de media móvil estacional y especifica cuántos términos de error pasados estacionales deben considerarse en la parte de media móvil estacional.

El parámetro $s$ en el modelo SARIMA representa la periodicidad de la estacionalidad y es específico para datos que exhiben patrones estacionales recurrentes a intervalos regulares. Este parámetro indica cuántas observaciones hay en cada temporada. Se refiere al número de períodos en una temporada estacional completa. Si los datos son diarios s=365,
si los datos son semanales s=52, si los datos mensuales s=12, etc.

En la practica $d=D$ y se busca la combinación de los parametros $P$ ,$Q$ que minimicen el AIC del modelo

```python
model = sm.tsa.statespace.SARIMAX(df1['Log_AveragePrice'],order = (2,1,2),seasonal_order = (0,1,0,52)) 
### DEBIDO A QUE LOS DATOS SON SEMANALES
model_fit = model.fit()
print(model_fit.summary()) ###
```

#### SUAVIZAMIENTO EXPONENCIAL SIMPLE

El suavizamiento exponencial es una técnica de pronóstico utilizada en análisis de series temporales y estadísticas para prever valores futuros en función de patrones observados en datos históricos. Esta técnica es especialmente útil cuando se trata de datos que muestran tendencias y patrones de estacionalidad.

El suavizamiento exponencial asigna un peso exponencialmente decreciente a las observaciones más antiguas, dando mayor importancia a las observaciones más recientes. Esto significa que los datos más recientes tienen un impacto más significativo en la predicción que los datos más antiguos. 


$\hat{Y}_{t+1} = \alpha \cdot Y_t + (1 - \alpha) \cdot \hat{Y}_t $


 Cuando 
α es alto, significa que se da más peso a las observaciones más recientes en la serie temporal, y se le resta importancia a las observaciones más antiguas.


si $\alpha =0$ entonces $\hat{Y}_{t+1} = \hat{Y}_t $ Como utilizamos las observaciones pasadas para predecir  $\hat{Y}$ entonces le damos mas importancia a las observaciones anteriores

si $\alpha =1$ entonces $\hat{Y}_{t+1} =  Y_t$ en este caso no miramos el pronostico que ha sido estimado con valores pasados y utilizamos solo los valores reales mas recientes

```python
# ajuste de parametros
list_smoothing=[]
list_rmse = []
list_aic = []
list_mae = []
modelos_exp =pd.DataFrame()
for i in np.arange(0.1,1,.1):

    mod_exp = SimpleExpSmoothing(train).fit(smoothing_level=round(i,1), optimized=False)
    fc = mod_exp.forecast(len(test))
    rmse = np.sqrt(mean_squared_error(test, fc.tolist()))
    mae = mean_absolute_error(test, fc.tolist())

    list_aic.append(mod_exp.aic)
    list_smoothing.append(i)
    list_rmse.append(rmse)
    list_mae.append(mae)
modelos_exp["smoothing_level"]=list_smoothing
modelos_exp["rmse"]=list_rmse
modelos_exp["aic"]=list_aic
modelos_exp["mae"]=list_mae

modelos_exp.sort_values(by="aic")
```

### Modelo de holt

El método de Holt, también conocido como suavizamiento exponencial doble o suavizamiento de nivel y tendencia, es una técnica utilizada para prever series temporales que muestran tanto una tendencia como una estacionalidad. Fue propuesto por Charles C. Holt en la década de 1950 y es una extensión del suavizamiento exponencial simple.

El modelo de Holt incorpora dos componentes principales: 

- el suavizamiento exponencial para la tendencia (smoothing_level) Un valor más cercano a 1 asignará más peso a las observaciones más recientes, lo que hará que el modelo sea más reactivo a los cambios recientes. Un valor más cercano a 0 dará menos peso a las observaciones recientes, lo que hará que el modelo sea más estable.

- el suavizamiento exponencial para el nivel (smoothing_slope). Estos componentes se combinan para prever la serie temporal y se adaptan para manejar tanto cambios en la tendencia como variaciones estacionales.  Un valor más cercano a 1 asignará más peso a las observaciones más recientes para la pendiente, lo que hará que el modelo sea más reactivo a los cambios recientes en la tendencia. Un valor más cercano a 0 dará menos peso a las observaciones recientes para la pendiente, haciendo que el modelo sea más estable.

```python
# ajuste de parametros
list_smoothing_level=[]
list_smoothing_slope=[]
list_rmse = []
list_aic = []
list_mae = []
modelos_holt =pd.DataFrame()
for i in np.arange(0.1,1,.1):
    for j in np.arange(0.1,1,.1):

        mod_holt = Holt(train).fit(smoothing_level=round(i,1),smoothing_slope=round(j,1))
        fc = mod_holt.forecast(len(test))
        rmse = np.sqrt(mean_squared_error(test, fc.tolist()))
        mae= mean_absolute_error(test, fc.tolist())
        


        list_smoothing_level.append(i)
        list_smoothing_slope.append(j)
        list_aic.append(mod_holt.aic)
        list_rmse.append(rmse)
        list_mae.append(mae)
modelos_holt["smoothing_level"]=list_smoothing_level
modelos_holt["smoothing_slope"]=list_smoothing_slope
modelos_holt["rmse"]=list_rmse
modelos_holt["aic"]=list_aic
modelos_holt["mae"]=list_mae
modelos_holt.sort_values(by="aic")
```

```python
model1 = Holt(train).fit(smoothing_level=0.9,smoothing_slope=.1)
fc1 = model1.forecast(len(test)).rename("smoothing_level = .9,smoothing_slope=.1")
model2 = Holt(train).fit(smoothing_level=0.8,smoothing_slope=.1)
fc2 = model2.forecast(len(test)).rename("smoothing_level = .8,smoothing_slope=.1")


plt.figure(figsize=(12, 8))
plt.plot(train, marker="o", color="black")
plt.plot(model1.fittedvalues, color="blue")
(line1,) = plt.plot(fc1, marker="o", color="blue")
plt.plot(model2.fittedvalues, color="red")
(line2,) = plt.plot(fc2, marker="o", color="red")

plt.legend([line1, line2], [fc1.name, fc2.name])
```

vemos como el modelo explico la tendencia

#### holt winters estacional

El modelo de suavizamiento exponencial con `trend="add"` y `seasonal="add"` es una variante del suavizamiento exponencial doble de Holt-Winters en el cual se asume que la tendencia y la estacionalidad son aditivas.

Aquí está el significado de los parámetros:

- **`trend="add"`:** Indica que la componente de tendencia se modela de manera aditiva. Esto significa que se asume que la tendencia en la serie temporal sigue una función lineal aditiva, es decir, la tendencia cambia en una cantidad constante con el tiempo.

- **`seasonal="add"`:** Indica que la componente estacional se modela de manera aditiva. Esto significa que se asume que los efectos estacionales siguen patrones que se suman constantemente a la serie temporal.

En resumen, con `trend="add"` y `seasonal="add"`, el modelo asume que la tendencia y la estacionalidad tienen efectos lineales que se suman a la serie temporal.

```python
model1 = ExponentialSmoothing(train,trend="add").fit()
fc1 = model1.forecast(len(test)).rename("trend")

model2 = ExponentialSmoothing(train,trend="add",seasonal="add",seasonal_periods=12).fit()
fc2 = model2.forecast(len(test)).rename("trend,seasonal")



plt.figure(figsize=(12, 8))
plt.plot(train, marker="o", color="black")
plt.plot(model1.fittedvalues, marker="o", color="blue")
(line1,) = plt.plot(fc1, marker="o", color="blue")
plt.plot(model2.fittedvalues, marker="o", color="red")
(line2,) = plt.plot(fc2, marker="o", color="red")

plt.legend([line1, line2], [fc1.name, fc2.name])
```
