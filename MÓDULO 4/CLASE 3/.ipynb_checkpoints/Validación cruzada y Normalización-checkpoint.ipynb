La validación cruzada (cross-validation) es una técnica utilizada en el aprendizaje automático para evaluar el rendimiento de un modelo predictivo y reducir el riesgo de overfitting . La idea fundamental detrás de la validación cruzada es dividir el conjunto de datos en varias partes, realizar el entrenamiento del modelo en algunas de esas partes y evaluar su rendimiento en las partes restantes (esta división la hacemos dentro del conjunto de entrenamiento

El proceso típico de validación cruzada implica los siguientes pasos:

- División del conjunto de entrenamiento : El conjunto de datos de entrenamiento se divide en $k$ subconjuntos (o "folds") de aproximadamente igual tamaño.

- Iteración: Se repite el siguiente proceso $k$ veces:

Se selecciona uno de los $k$ subconjuntos como conjunto de prueba.

Los otros $k-1$ subconjuntos se utilizan como conjunto de entrenamiento.

- Entrenamiento y evaluación: El modelo se entrena en el conjunto de entrenamiento y se evalúa en el conjunto de prueba. Se registra la métrica de rendimiento de nuestra elección (r2,accuracy, mse,etc)

- Promedio de resultados: Se promedian los resultados obtenidos en las $k$ iteraciones para obtener una estimación más robusta del rendimiento del modelo.

La validación cruzada ayuda a mitigar la variabilidad en la estimación del rendimiento del modelo que puede surgir debido a una partición específica de los datos. Esto es especialmente útil cuando se tiene un conjunto de datos limitado y se busca una evaluación más precisa del rendimiento del modelo.

### Ejemplo con el dataset de INEGI

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV,cross_val_score,cross_validate
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from sklearn.metrics import make_scorer
```

```python
df = pd.read_csv("ingresohogaresinegi.csv") #cargamos dataset
```

```python
df.columns # vemos columnas
```

```python
df = df[['Alf', 'Asi', 'EBC', 'Der', 'Pis',
       'San', 'Agu', 'Dre', 'Ele', 'Lav', 'Ref', 'Ing']] # nos quedamos unicamente las columnas numericas
```

```python
X= df.drop(columns='Ing') 
Y= np.log(df['Ing']) #transformamos la variable Y con logaritmo
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.3, random_state=0) # partimos 70% entrenamiento, 30# prueba
```

ajustamos un modelo lineal, utilizamos una semilla y una proporcion de entrenamiento distinta a la del ejemplo de la clase pasada para ver como cambian los valores de r2

```python
lin_reg =  LinearRegression().fit(X_train,Y_train)
y_pred = lin_reg.predict(X_test)
y_pred_train =  lin_reg.predict(X_train)
print(np.round(lin_reg.score(X_train, Y_train),5))
print(np.round(lin_reg.score(X_test, Y_test),5))
```

si bien no tenemos la mejor r2 no se ven diferencias que muestren overfitting. Pero realmente el desempeño del modelo es de .50712 ?

#### usamos validación cruzada

```python
scores=cross_validate(lin_reg,X_train,Y_train,cv=3,scoring="r2",return_train_score=True)
scores # dividimos en 3 particiones (folds) y calculamos la r2
```

partimos el conjunto de entrenamiento en 3 y ajustamos el mismo modelo. Podemos ver los resultados de r2 para train y test en cada una de las particiones

```python
test_score_promedio=scores["test_score"].mean()
train_score_promedio=scores["train_score"].mean()
print(train_score_promedio)
print(test_score_promedio)
```

Promediando los resultados,  vemos que el valor de r2 se aleja en .08 del que encontramos en el conjunto de test

Una de las finalidades de la validación cruzada es poder ajustar los hiperparámetros del modelo por medio de los resultados obtenidos. Con la función $ gridsearchCV $ podemos hacer esta tarea. Incluso podemos seleccionar varios modelos distintos con varias combinaciones de hiperparámetros y la funcion gridsearchCV nos devolverá el modelo con los parámetros que tuvieron mejores resultados en la validación cruzada

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.3, random_state=0)
```

```python
#my_score = make_scorer(r2_score, greater_is_better = True)
```

evaluaremos el modelo clasido de regresion,lasso, ridge y elastic net con distintos valores de alpha

```python
model_params={
    "reg_lasso":{
        "model":Lasso(), # modelo lasso
        "params":{
            "alpha":[i for i in np.arange(.001,2,.005)] # valores entre .001 y 2 (dando saltos de .005)

        }
    },
    
    "reg_ridge":{
        "model":Ridge(), #modelo ridge
        "params":{
            "alpha":[i for i in np.arange(.001,2,.005)]# valores entre .001 y 2 (dando saltos de .005)

        }
    },
    
    "elastic_net":{
        "model":ElasticNet(), #elastic net
        "params":{
            "alpha":[i for i in np.arange(.001,2,.005)],# valores entre .001 y 2 (dando saltos de .005)
            "l1_ratio":[i for i in np.arange(.1,1,.1)] # valores entre .1 y 1 (dando saltos de .1)

        }
    },
    
    "reg_lin":{
        "model":LinearRegression(), #Regresion lineal
        "params":{
            'fit_intercept': [True,False] # modelo con y sin intercepto (coeficiente Bo)
        }
    }
            
}
```

```python
r2_models=[] # lista para guardar el r2 de cada uno de los modelos
for model_name,mp in model_params.items():

    modelo=GridSearchCV(mp["model"],mp["params"],cv=3,n_jobs=-1,scoring="r2",verbose=0) #evaluamos con r2 y partimos en 3 el conjunto de train
    
    modelo.fit(X_train,Y_train)
    r2_models.append({
        "model":model_name,
        "best_score":modelo.best_score_,
        "best_params":modelo.best_params_
    })
```

```python
### resumimos en una tabla los resultados obtenidos en gridsearch
df_r2_models=pd.DataFrame(r2_models).sort_values(by="best_score",ascending=False)
df_r2_models
```

el mejor modelo fue Ridge con r2 de .46 y con un valor de alpha de .05

 Confirmamos de manera individual los resultados

```python
param_grid = {"alpha":[.051]
              
              
             }  
   
grid = GridSearchCV(Ridge(), param_grid, refit = True, cv = 3,verbose=0,scoring="r2") 
   

grid.fit(X_train, Y_train) 
 
# vemos el resumen de gridsearch
grid.cv_results_
```

Ajustamos el modelo ganador

```python
lin_reg =  Ridge(alpha=0.051).fit(X_train,Y_train)
y_pred = lin_reg.predict(X_test)
y_pred_train =  lin_reg.predict(X_train)
print("r2 train",np.round(lin_reg.score(X_train, Y_train),5))
print("r2 test",np.round(lin_reg.score(X_test, Y_test),5))
```

podemos reportar el r2 = .49 como el score final de nuestro modelo. Vemos que no hay una diferencia que muestre overfitting

# NORMALIZACION Y ESTANDARIZACION

La normalización y estandarización son técnicas comunes de preprocesamiento de datos en machine learning. Ambas tienen como objetivo ajustar las características de los datos para que sean más adecuadas para los algoritmos de aprendizaje automático. Aquí hay una breve definición de ambas:

Normalización: La normalización se refiere al proceso de escalar y centrar los datos en un rango específico, generalmente entre 0 y 1. 

$$  Xnorm = \frac{X - min(X)}{ max(X) - min(X)} $$



Estandarización: La estandarización implica transformar los datos para que tengan una media de cero y una desviación estándar de uno. Esto se logra restando la media y dividiendo por la desviación estándar.

$$  Xestandard = \frac{X - mean(X)}{std(X)} $$

la documentación de [StandardScaler](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html) dice (Traducido al español):

>*La estandarización de un conjunto de datos es un requisito común para muchos estimadores de aprendizaje automático: podrían comportarse mal si las características individuales no se parecen más o menos a datos normalmente distribuidos estándar (por ejemplo, gaussianos con media 0 y varianza unitaria)..*

>*Por ejemplo, muchos elementos utilizados en la función objetivo de un algoritmo de aprendizaje  suponen que todas las características están centradas alrededor de 0 y tienen una varianza del mismo orden. Si una característica tiene una varianza que es órdenes de magnitud mayor que otras, puede dominar la función objetivo y hacer que el estimador no pueda aprender de otras características correctamente como se espera.*

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.datasets import make_regression
import pandas as pd

# Generar un conjunto de datos de regresión de ejemplo
X, y = make_regression(n_samples=100, n_features=3, noise=0.1, random_state=42)

# Convertir a un DataFrame para facilitar la visualización
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2', 'Feature_3'])
df['Target'] = y

# Mostrar el conjunto de datos original
print("Conjunto de datos original:")
print(df.head())

# Normalización
scaler_minmax = MinMaxScaler()
X_normalized = scaler_minmax.fit_transform(X)

# Estandarización
scaler_standard = StandardScaler()
X_standardized = scaler_standard.fit_transform(X)

# Convertir a DataFrames para facilitar la visualización
df_normalized = pd.DataFrame(X_normalized, columns=['Feature_1', 'Feature_2', 'Feature_3'])
df_standardized = pd.DataFrame(X_standardized, columns=['Feature_1', 'Feature_2', 'Feature_3'])

# Mostrar conjuntos de datos transformados
print("\nConjunto de datos normalizado:")
print(df_normalized.head())

print("\nConjunto de datos estandarizado:")
print(df_standardized.head())
```

### DE MANERA MANUAL

```python
X, y = make_regression(n_samples=100, n_features=3, noise=0.1, random_state=42)

# Convertir a un DataFrame para facilitar la visualización
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2', 'Feature_3'])
df['Target'] = y

# Mostrar el conjunto de datos original
print("Conjunto de datos original:")
print(df.head())

# Normalización
df_normal = df.copy()
for i in df.drop(columns="Target").columns:
    df_normal[i]= (df_normal[i] - min(df_normal[i])) / (max(df_normal[i])-min(df_normal[i]))
    

# Estandarización
df_estandar = df.copy()
for i in df.drop(columns="Target").columns:
    df_estandar[i]= (df_estandar[i] - df_estandar[i].mean()) / df_estandar[i].std()

# Convertir a DataFrames para facilitar la visualización


# Mostrar conjuntos de datos transformados
print("\nConjunto de datos normalizado:")
print(df_normal.drop(columns="Target").head())

print("\nConjunto de datos estandarizado:")
print(df_estandar.drop(columns="Target").head())
```
