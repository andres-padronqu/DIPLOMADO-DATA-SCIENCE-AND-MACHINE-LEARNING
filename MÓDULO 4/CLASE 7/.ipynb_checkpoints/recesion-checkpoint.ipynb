Los bonos del Tesoro de Estados Unidos, también conocidos como bonos del gobierno de EE. UU. o simplemente "Treasuries" (Tesorerías), son instrumentos de deuda emitidos por el Departamento del Tesoro de los Estados Unidos. Estos bonos son una forma en la que el gobierno federal puede pedir prestado dinero del público y de otras instituciones para financiar sus gastos y obligaciones.

Aquí hay algunas características clave de los bonos del Tesoro de Estados Unidos:

Garantía del Gobierno Federal: Los bonos del Tesoro son considerados activos libres de riesgo, ya que están respaldados por la capacidad del gobierno de EE. UU. para recaudar impuestos y emitir moneda. Esto los convierte en uno de los activos más seguros disponibles en los mercados financieros.

Varios Vencimientos: Los bonos del Tesoro se emiten con una variedad de vencimientos, que pueden ir desde tan cortos como unos pocos días hasta tan largos como 30 años. Algunos ejemplos incluyen bonos a corto plazo (como el Bonos del Tesoro a 2 años), bonos a mediano plazo (como el Bonos del Tesoro a 10 años) y bonos a largo plazo (como el Bonos del Tesoro a 30 años).

Pagos de Intereses: Los bonos del Tesoro pagan intereses regularmente a los tenedores de bonos. Los intereses se calculan sobre el valor nominal del bono y se pagan en intervalos predeterminados.

Inversión y Compra en Subastas: Los inversores pueden comprar bonos del Tesoro directamente en subastas organizadas por el Departamento del Tesoro. También pueden adquirirse en el mercado secundario a través de intermediarios financieros.

Liquidez: Los bonos del Tesoro son conocidos por su alta liquidez, lo que significa que es relativamente fácil comprar y vender estos bonos en el mercado secundario.

Uso en Estrategias de Inversión: Dada su naturaleza segura y su papel como referencia para las tasas de interés libres de riesgo, los bonos del Tesoro son utilizados en diversas estrategias de inversión y sirven como indicadores de las condiciones del mercado financiero.

El Departamento del Tesoro emite estos bonos para financiar el déficit presupuestario del gobierno y satisfacer otras necesidades de endeudamiento. La emisión y gestión de los bonos del Tesoro son responsabilidad del Bureau of the Fiscal Service del Departamento del Tesoro.






Las abreviaturas "DGS2", "DGS10" y "DGS30" hacen referencia a tasas de rendimiento (o rendimientos) de bonos del Tesoro de los Estados Unidos con diferentes vencimientos. Estas tasas representan los rendimientos de los bonos del Tesoro con vencimientos de 2, 10 y 30 años, respectivamente.

DGS2: Tasa de rendimiento del bono del Tesoro de Estados Unidos con un vencimiento de 2 años.

DGS10: Tasa de rendimiento del bono del Tesoro de Estados Unidos con un vencimiento de 10 años.

DGS30: Tasa de rendimiento del bono del Tesoro de Estados Unidos con un vencimiento de 30 años.

Estas tasas de rendimiento son indicadores importantes en los mercados financieros y son utilizadas para evaluar las condiciones económicas y las expectativas de los inversionistas. Por lo general, las tasas de rendimiento de bonos a corto plazo (como el bono de 2 años) tienden a ser más sensibles a los cambios en las tasas de interés de la Reserva Federal, mientras que las tasas de rendimiento a largo plazo (como el bono de 30 años) pueden reflejar las expectativas económicas a más largo plazo y la inflación.

Los datos de estas tasas de rendimiento son comúnmente proporcionados por fuentes financieras y están disponibles en tiempo real. Pueden ser utilizados por analistas, inversionistas y economistas para realizar análisis de la curva de rendimiento, evaluar las perspectivas económicas y tomar decisiones de inversión.



La variable "USRECD" se refiere a la recesión en Estados Unidos y es a menudo utilizada como un indicador económico. En muchos conjuntos de datos económicos, esta variable binaria se utiliza para señalar si la economía de Estados Unidos está en recesión o no en un período de tiempo específico.

La variable "USRECD" generalmente toma uno de los siguientes valores:

1: Indica que la economía de Estados Unidos está en recesión durante ese período.

0: Indica que la economía de Estados Unidos no está en recesión durante ese período.

Esta variable puede estar basada en diversos indicadores económicos, como el crecimiento del Producto Interno Bruto (PIB), el empleo, la producción industrial, entre otros. La determinación de si la economía está en recesión suele seguir criterios definidos, como dos trimestres consecutivos de contracción en el PIB, aunque los métodos específicos pueden variar.

La utilización de la variable "USRECD" permite a los analistas y economistas estudiar y analizar patrones económicos a lo largo del tiempo y comprender mejor la dinámica de las expansiones y contracciones económicas en Estados Unidos. Este tipo de datos es valioso para la toma de decisiones en inversiones, políticas económicas y planificación financiera.

Una recesión es una fase del ciclo económico caracterizada por una disminución generalizada de la actividad económica, que se manifiesta en una contracción de diversos indicadores económicos durante un período prolongado. En términos generales, se considera que una economía está en recesión cuando el Producto Interno Bruto (PIB) disminuye durante al menos dos trimestres consecutivos.

Aquí hay algunas características clave de una recesión:

Contracción Económica: Durante una recesión, la producción económica, medida por el PIB, disminuye. Las empresas pueden reducir la producción, las ventas disminuyen, y la inversión y el gasto del consumidor suelen disminuir.

Aumento del Desempleo: La recesión a menudo se asocia con un aumento del desempleo, ya que las empresas reducen la contratación o incluso recortan empleos para enfrentar la disminución de la demanda.

Caída en la Inversión: Las empresas pueden posponer o cancelar inversiones en nuevos proyectos durante una recesión, ya que enfrentan una mayor incertidumbre y condiciones financieras más desafiantes.

Reducción en el Gasto del Consumidor: Durante las recesiones, los consumidores pueden reducir el gasto, especialmente en bienes y servicios no esenciales. Esta disminución en el consumo puede agravar la contracción económica.

Disminución en la Producción Industrial: La producción industrial suele disminuir durante las recesiones, ya que las empresas reducen la fabricación de bienes debido a la menor demanda.

Baja en los Ingresos y Utilidades: La recesión puede estar acompañada de una disminución en los ingresos y las ganancias de las empresas, lo que afecta negativamente la salud financiera de las empresas.

Política Monetaria y Fiscal: Durante las recesiones, los gobiernos y los bancos centrales a menudo implementan políticas para estimular la economía, como reducciones en las tasas de interés, programas de estímulo fiscal o medidas para respaldar el sistema financiero.

La diferencia entre las tasas de rendimiento de bonos a 10 años (DGS10) y bonos a 2 años (DGS2), comúnmente conocida como "diferencial de la curva de rendimiento" o "spread de la curva de rendimiento", es un indicador financiero que ha sido históricamente observado por analistas y economistas como una señal potencial de recesión económica. Este diferencial es relevante porque refleja la estructura temporal de las tasas de interés y puede proporcionar información sobre las expectativas del mercado en cuanto a la dirección de la economía.

La teoría detrás del uso del diferencial de la curva de rendimiento para estimar recesiones se basa en la idea de que, en condiciones normales, las tasas de interés a largo plazo tienden a ser más altas que las tasas a corto plazo debido al riesgo temporal y a la incertidumbre futura. Cuando la economía está en expansión, los inversores suelen exigir tasas de interés más altas para comprometer su dinero durante un período más largo.

Sin embargo, en anticipación a una posible recesión, los inversores pueden volverse más cautelosos y buscar refugio en inversiones a corto plazo, lo que eleva las tasas de interés a corto plazo en comparación con las tasas a largo plazo. Esta inversión en la curva de rendimiento puede llevar a un diferencial negativo o reducir el diferencial positivo.

¿Por qué puede ser útil para estimar recesiones?
Historia como indicador adelantado: En el pasado, inversiones de la curva de rendimiento (inversión del diferencial) han precedido a muchas recesiones económicas. Cuando el diferencial se invierte (las tasas a corto plazo superan a las tasas a largo plazo), ha habido casos históricos en los que esto ha sido seguido por recesiones.

Expectativas del mercado: El diferencial de la curva de rendimiento refleja las expectativas del mercado sobre la dirección futura de la economía. Los inversionistas pueden interpretar una inversión como una señal de que el mercado espera condiciones económicas más difíciles en el futuro.

Política monetaria y expectativas de inflación: Los cambios en el diferencial también pueden relacionarse con cambios en la política monetaria y las expectativas de inflación. Estos factores pueden influir en las decisiones de los inversores y en la evaluación del riesgo económico.

Es importante tener en cuenta que aunque el diferencial de la curva de rendimiento ha sido un indicador útil en el pasado, no garantiza con certeza la ocurrencia de una recesión y puede haber otras variables económicas a considerar. Además, el análisis de la curva de rendimiento se basa en patrones históricos y no siempre se repiten en todas las situaciones económicas.

```python
recesion2 = recesion2.iloc[:-250]
```

```python
logreg=Pipeline([
    ("mean_imp",SimpleImputer()),
    ("log_reg",LogisticRegression(fit_intercept=False))
])
```

```python
xtrain,ytrain=recesion2.Spread.values.reshape(-1,1), recesion2.USRECD.values
```

```python
logreg.fit(xtrain,ytrain)
```

```python
recesion2.plot(x="Spread",y="USRECD",kind="scatter",alpha=0.4)
xrange=recesion2.Spread.values.reshape(-1,1)
plt.plot(xrange,logreg.predict_proba(xrange)[:,1],
        color="tab:orange") ### LA PROBABILIDAD DE RECESION ES MAYOR EN SPREADS NEGATIVOS
```

```python
prueba1=recesion1[["Spread","USRECD"]]
#RECESION 1 trae todos los datos sin rezagar, trae los datos de hoy,
```

```python
probas=logreg.predict_proba(prueba1[["Spread"]].values.reshape(-1,1))
probas1=pd.DataFrame( [i[1] for i in probas] ,columns=["Probas"]).set_index(recesion1.index)
probas1
```

```python
probas1=probas1.join(recesion1)
```
