Un árbol de decisión es un modelo de aprendizaje supervisado utilizado en machine learning y data science. Este modelo es representado en forma de un árbol, donde cada nodo interno representa una decisión basada en una característica específica, cada rama representa el resultado de la decisión, y cada hoja del árbol representa la etiqueta de la clase o el valor de la variable que estamos tratando de predecir. Este modelo se puede utilizar tanto en problemas de clasificación como en problemas de regresión

En el contexto de detección de fraude podríamos tener para este conjunto de datos un arbol como el siguiente
<img src="muestradatos.png">
<img src="arbol1.png">

La estructura de un arbol de decisión es la siguiente:


#### Nodo Raíz:

El nodo raíz es el primer nodo en un árbol de decisión. Es el punto de partida desde el cual se inicia la construcción del árbol. Este Representa el conjunto completo de datos en el que se basa la toma de decisiones. Se divide en ramas, cada una asociada a una condición sobre una característica. Este nodo apunta a los nodos internos pero ningun nodo apunta a él

#### Nodo Interno:

Los nodos internos son aquellos que se encuentran entre el nodo raíz y los nodos hojas. Son puntos de decisión que dividen el conjunto de datos en subconjuntos más pequeños. Representan una condición sobre una característica que guía la división de datos.Cada rama que sale de un nodo interno está asociada a una respuesta "verdadera" o "falsa" para la condición especificada. Estos nodos apuntan a otros y tambien son apuntados por nodos superiores

### Nodo Hoja:

Las hojas son los nodos terminales del árbol de decisión. No tienen ramas salientes y representan la predicción final o la etiqueta de clase para un subconjunto específico de datos. Estos nodos no apuntan a ninguno otro.

Necesitamos ahora cuantificar la impureza de cada nodo para los arboles que creamos. Esto lo hacemos con el indice de Gini que se define como sigue:

$\gamma_s=1-\sum_{j=0}^1 \hat{p}_s(j)^2$.

donde 

$ \hat{p}_s(j)=\dfrac{\sum_{i=1}^n I(Y_i=j,X_i\in A_s)}{\sum_{i=1}^n I(X_i\in A_s)} $

con $j=0,1$.


 Si una partición $A_s$ contiene todos los 0's o 1's entonces $\gamma_s=0$. De otra manera $\gamma_s>0$ Entonces nosotros tratamos de escoger un punto $t$ tal que minimize la impureza.

Con lo anterior podemos concluir que el proceso de construcción de un arbol de decisión es el siguiente



#### Selección del atributo de decisión (característica): 

En cada nodo del árbol, se selecciona la característica que mejor divide los datos en subconjuntos más puros. La "pureza" se refiere a la homogeneidad de las muestras en cada subconjunto. El objetivo es maximizar la homogeneidad y minimizar la impureza.

#### Criterio de división: 

Para medir la pureza de un nodo, se utilizan diferentes criterios, como Gini impurity, entropía o error de clasificación. El algoritmo selecciona la característica y el punto de corte que minimizan la impureza.

#### División de los datos: 

Una vez seleccionada la característica y el punto de corte, se dividen los datos en dos conjuntos en función de la condición del nodo. Este proceso se repite recursivamente para cada subconjunto hasta alcanzar un criterio de parada, como la profundidad máxima del árbol.

#### Asignación de etiquetas a las hojas: 

Cuando se completa la construcción del árbol, cada hoja se etiqueta con la clase o el valor más frecuente en el conjunto de datos correspondiente a esa hoja.

#### Poda (opcional): 

Después de construir el árbol, se puede realizar un proceso de poda para evitar el sobreajuste (overfitting) al conjunto de entrenamiento. La poda implica eliminar ramas del árbol que no contribuyen significativamente a la mejora de la predicción en el conjunto de prueba.

#### TIPOS DE ARBOLES DE DECISION


● ID3: A Ross Quinlan se le atribuye el desarrollo de ID3 en 1986, que es la abreviatura de "Iterative
Dichotomized 3". Este algoritmo aprovecha la entropía definida como 

y la ganancia de información como métricas para evaluar las divisiones de candidatos.

● C4.5: Este algoritmo se considera una iteración posterior de ID3, que también fue desarrollado por
Quinlan. Puede utilizar la ganancia de información o las proporciones de ganancia para evaluar los
puntos de división dentro de los árboles de decisión.

● CART: El término CART es una abreviatura de "árboles de clasificación y regresión" (“classification
and regression trees”) y fue introducido por Leo Breiman. Este algoritmo generalmente utiliza la
impureza de Gini para identificar el atributo ideal para la división. La impureza de Gini mide la
frecuencia con la que se clasifica incorrectamente un atributo elegido al azar. Cuando se evalúa
usando la impureza de Gini, un valor más bajo es más ideal.


Los arboles que construimos anteriormente fueron usando el algoritmo CART


Como todos los modelos de ML tenemos ventajas y desventajas:

### Ventajas

● Fácil de interpretar: La lógica booleana y las representaciones visuales de los árboles de
decisión los hacen más fáciles de entender y consumir. también facilita ver qué atributos son los más importantes, lo que en algunos modelos puede no ser claro.

● Se requiere poca o ninguna preparación de datos: Los árboles de decisión tienen una serie
de características que los hacen más flexibles que otros clasificadores. Puede manejar varios
tipos de datos, es decir, valores discretos o continuos. 

● Más flexible: Los árboles de decisión se pueden aprovechar para tareas de clasificación y
regresión, lo que los hace más flexibles que otros algoritmos. También es insensible a las
relaciones subyacentes entre los atributos; esto significa que si dos variables están altamente
correlacionadas, el algoritmo sólo elegirá una de las características para realizar la división.

    
### Desventajas:
● Propenso al sobreajuste: Los árboles de decisión complejos tienden a sobreajustarse y no se
generalizan bien a los nuevos datos. Este escenario se puede evitar mediante los procesos de
poda previa o poda posterior. La poda previa detiene el crecimiento del árbol cuando no hay
datos suficientes, mientras que la poda posterior elimina los subárboles con datos inadecuados
después de la construcción del árbol.

● Estimadores de alta varianza: Pequeñas variaciones dentro de los datos pueden producir un
árbol de decisión muy diferente. Bagging y Boosting, o el promedio de estimaciones, puede ser un
método para reducir la varianza de los árboles de decisión.

● Más costoso: Dado que los árboles de decisión adoptan un enfoque de búsqueda codicioso
durante la construcción, pueden ser más costosos de entrenar en comparación con otros
algoritmos.

```python
import numpy as np
import pandas as pd
import requests
from io import StringIO

from sklearn.model_selection import train_test_split

from sklearn.tree import (DecisionTreeClassifier, 
                         DecisionTreeRegressor,
                         plot_tree,
                         export_graphviz)
import matplotlib.pyplot as plt

from sklearn.model_selection import cross_validate
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import confusion_matrix
import seaborn as sns
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import precision_score,accuracy_score
from sklearn.metrics import roc_curve, roc_auc_score
from sklearn.model_selection import KFold
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import make_scorer
from sklearn.metrics import mean_squared_error
from sklearn.ensemble import AdaBoostClassifier,AdaBoostRegressor,RandomForestRegressor,RandomForestClassifier
from xgboost import  XGBClassifier,XGBRegressor
```

```python
datalink="http://archive.ics.uci.edu/ml/machine-learning-databases/mammographic-masses/mammographic_masses.data"
df=pd.read_csv(StringIO(requests.get(datalink).text),na_values="?",header=None)
df.columns=["bi_rads","age","shape","margin","density","severity"]
df=df.drop(columns="bi_rads")
df["age"] = df["age"].fillna(df["age"].mean())
df["shape"] = df["shape"].fillna(4)
df["margin"] = df["margin"].fillna(1)
df["density"] = df["density"].fillna(3)
### convertimos a variables dummies
dummies_shape=pd.get_dummies(df["shape"],prefix="shape",drop_first=True).replace(True,1).replace(False,0)
dummies_margin=pd.get_dummies(df["margin"],prefix="margin",drop_first=True).replace(True,1).replace(False,0)
dummies_density=pd.get_dummies(df["density"],prefix="density",drop_first=True).replace(True,1).replace(False,0)
df = pd.concat([df,dummies_shape,dummies_margin,dummies_density],axis=1)
df=df.drop(columns=["shape","margin","density"])
X=df.drop(columns="severity")
Y=df["severity"]
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.3, random_state=3)
```

La poda es una técnica que se utiliza para reducir el sobreajuste. La poda también simplifica un árbol de decisiones al eliminar las reglas más débiles. Básicamente, la poda recursiva encuentra el nodo con el "eslabón más débil". El eslabón más débil se caracteriza por un alfa efectivo, donde los nodos con el alfa efectivo más pequeño se podan primero.
$$
R_\alpha(T)=R(T)+\alpha |T|
$$

$R(T)$= error total de entrenamiento en las hojas
$\alpha$= parametro de complejidad
$|T|$= numero de hojas en el arbol

<img src="poda_arbol.png">


DecisionTreeClassifier proporciona parámetros como $\textit{min_samples_leaf}$ y $\textit{max_depth}$ para evitar que un árbol se sobreajuste. La poda de costo complejidad proporciona otra opción para controlar el tamaño de un árbol. En DecisionTreeClassifier, esta técnica de poda está parametrizada por el parámetro de complejidad de costo, $\textit{ccp_alpha}$.



```python
path = tree_clf.cost_complexity_pruning_path(X_train, Y_train)
#print(path)
ccp_alphas, impurities = path.ccp_alphas, path.impurities
print(ccp_alphas)
```

```python
#### usamos los valores de ccp_alpha para entrenar el modelo y los guardamos en una lista
models = []
for ccp_alpha in ccp_alphas:
    mod = DecisionTreeClassifier(ccp_alpha=ccp_alpha,random_state=0)
    mod.fit(X_train, Y_train)
    models.append(mod)
print("Numero de nodos en el ultimo arbol es: {} con ccp_alpha: {}".format(
      models[-1].tree_.node_count, ccp_alphas[-1]))
models
```

```python
print(models[4].tree_.node_count)
print(models[4].tree_.max_depth) # el cuarto modelo con ccp_alpha=.013 tiene 5 nodos y profundidad 2
```

```python
### eliminamos el ultimo caso ya que es un modelo con un solo nodo
models = models[:-1]
ccp_alphas = ccp_alphas[:-1]
```

```python
train_scores = [mod.score(X_train,Y_train) for mod in models]
test_scores = [mod.score(X_test, Y_test) for mod in models]

fig, ax = plt.subplots()
ax.set_xlabel("alpha")
ax.set_ylabel("accuracy")
ax.set_title("Accuracy vs alpha para datos de training y test")
ax.plot(ccp_alphas, train_scores, marker='o', label="train",
        drawstyle="steps-post")
ax.plot(ccp_alphas, test_scores, marker='o', label="test",
        drawstyle="steps-post")
ax.legend()
plt.show()
```

Con los valores iniciales más pequeños de alfa, existe una enorme disparidad entre las puntuaciones de precisión del entrenamiento y de las pruebas. Cuando el valor alfa aumenta, la precisión del entrenamiento disminuye y la precisión de la prueba aumenta hasta que alcanzamos el punto óptimo (puntuación más alta de precisión de la prueba). En el gráfico, el punto óptimo se encuentra en .013 de alfa, donde obtenemos el árbol podado óptimo

### Arboles de Regresión

Un árbol de regresión es un tipo de modelo de regresión no lineal que se utiliza en estadística y aprendizaje automático. A diferencia de los árboles de clasificación que se utilizan para problemas de clasificación, los árboles de regresión se emplean cuando la variable objetivo es continua en lugar de categórica.
Los árboles de regresión son flexibles y pueden capturar patrones no lineales en los datos.

###  La manera en que funciona un árbol de regresión es la siguiente:

1. **Dividir el conjunto de datos:** El algoritmo comienza dividiendo el conjunto de datos en subconjuntos más pequeños de tal forma que cada subconjunto sea lo más homogéneo posible con respecto a la variable objetivo. 


2. **Seleccionar la mejor división o 'split'**: Para cada subconjunto, el algoritmo busca la variable predictora que mejor separa los datos en dos grupos, en función de algún criterio (como la reducción del error cuadrático medio).


3. **Crear un nodo**: El algoritmo crea un nodo para cada subconjunto y le asigna la mejor predictora en función del split. El nodo representa un punto de decisión, donde los datos de entrada se dividen según la variable predictora que mejor separa y el valor del 'split'.


4. **Repetir:** Repite el proceso en cada subconjunto, creando nuevos nodos y dividiendo los datos aún más hasta que se alcance un criterio. Este puede ser el mínimo número de obs en una hoja, máxima profundidad del árbol, etc. 


5. **Predicciones:** Para hacer una predicción, los datos de entrada se alimentan por el árbol desde el nodo raíz hasta un nodo hoja, y se devuelve el valor objetivo correspondiente al nodo hoja como la predicción.

5. **Poda Opcional:**: Los arboles de regresión pueden sobreajustar facilmenteEstrategias como la poda del árbol (reducción de la profundidad del árbol) y el uso de conjuntos de árboles, como en el caso de Bosques Aleatorios, son comunes para mejorar la generalización del modelo.

Para el anterior conjunto de datos tenemos una relación que no es lineal. Un arbol de decisión como el anterior parece ser un buen modelo para estimar la efectividad de un medicamento con base en la cantidad de dosis que toma cada paciente. Nuevamente nos preguntamos. ¿Cómo elegimos la raiz del arbol?

#### Procedimiento
* 1: Promediamos los valores de las primeras dos observaciones del eje X, este valor lo llamaremos umbral
* 2: Dividimos el conjunto de datos en dos grupos, los que tienen su valor en el eje x a la izquierda y a la derecha del umbral
* 3: Sacamos el promedio de la variable respuesta Y para los dos grupos que creamos en el paso 2
* 4: El promedio Calculado en el paso 3 Será nuestra Y estimada ($\hat{y}$), un valor será estimado para el conjunto izquierdo y otro valor será estimado para el conjunto derecho
* 5: Calculamos el error $y_i - \hat{y_i}$ para cada una de las observaciones 
* 6: Calculamos el error del umbral como $\sum_{i=1}^n {y_i - \hat{y_i}}$
* 7: Repetimos todos los pasos anteriores para el siguiente par de observaciones en el eje X hasta que hayamos recorrido todas las observaciones

FINALMENTE escogemos como Raiz el umbral que minimaza el error $\sum_{i=1}^n {y_i - \hat{y_i}}$

<img src="ArbolRegresión2.PNG">

Para mas detalle de como calcular el anterior procedimiento, revisar el archivo $CalculoRaizArbolRegresion.xlsx$

Despues de calcular la raiz del arbol repetimos el procedimiento con los valores a la derecha y a la izquierda del umbral de la raiz, para crear las ramas con las siguientes decisiones, pero debemos tener cuidado ya que podemos caer en un caso como el siguiente

del umbral escogido para la raiz

# Ensemble Learning

El ensamble learning (Aprendizaje grupal) es una técnica que combina múltiples modelos de aprendizaje para mejorar el rendimiento y la generalización. Dos enfoques comunes dentro del ensamble learning son el bagging y el boosting.

#### Bagging (Bootstrap Aggregating):

Bagging consiste en entrenar múltiples modelos independiente s(usualmente arboles de decisión sin "podar") utilizando subconjuntos aleatorios (bootstrap) de los datos de entrenamiento.

Funciona generando múltiples conjuntos de datos de entrenamiento mediante muestreo con reemplazo, y cada modelo se entrena en uno de estos conjuntos. Luego, las predicciones de cada modelo se promedian (en el caso de regresión) o se toma la votación (en el caso de clasificación) para producir la predicción final.

Bagging ayuda a reducir la varianza y mejorar la estabilidad del modelo, especialmente cuando se utilizan modelos propensos al sobreajuste 

#### Boosting:

Boosting se centra en mejorar el rendimiento de modelos débiles, asignándoles pesos y ajustándolos iterativamente para corregir los errores cometidos en las predicciones anteriores.

Los modelos se construyen secuencialmente, y cada modelo posterior se enfoca en corregir los errores de los modelos anteriores. Se asignan pesos a las instancias mal clasificadas, de modo que el modelo siguiente preste más atención a esos casos en el proceso de entrenamiento.

Boosting es eficaz para mejorar la precisión del modelo, especialmente en casos en los que bagging no es suficiente. Los modelos débiles se combinan para formar un modelo fuerte que tiene un mejor rendimiento general.


<img src="baggingboosting.png">

Ambos enfoques, bagging y boosting, se utilizan para reducir el sesgo y la varianza, mejorar la capacidad de generalización y construir modelos más robustos y precisos. Algunos algoritmos populares que emplean estas técnicas incluyen Random Forest (bagging) y AdaBoost, Gradient Boosting Machine (GBM), y XGBoost (boosting).

### RANDOM FOREST (Bosque aleatorio)

El ejemplo mas famoso de bagging es probablemente el de _bosques aleatorios_ (**Random forests**), este modelo consiste en explicitamente construir arboles de decisión sin podar (strong learners) en muestras aleatorias con remplazo (bootstraping) y decidir nuestra predicción final apartir de cada uno de ellos

En general, mientras más árboles en el bosque se vean, más robusto es el bosque

El algoritmo consta de los siguientes pasos:

- 1 Selección aleatoria de muestras:

A partir del conjunto de datos original, se selecciona aleatoriamente un conjunto de muestras con reemplazo (bootstrap). Esto significa que algunas muestras pueden aparecer más de una vez en el conjunto de datos de entrenamiento para cada árbol individual.

- 2 Construcción de árboles de decisión:

Para cada conjunto de muestras seleccionado, se construye un árbol de decisión. Durante la construcción del árbol, en cada nodo se realiza una selección aleatoria de un subconjunto de características.

- 3 Votación o promedio:

Cuando se realiza una predicción para una nueva instancia, cada árbol en el bosque emite una predicción. Para problemas de clasificación, se realiza una votación entre los árboles y la clase más votada se elige como la predicción final. En problemas de regresión, se promedian las predicciones de todos los árboles.

- 4 Evaluación del rendimiento:

Se evalúa el rendimiento del modelo utilizando métricas relevantes como precisión, F1-score (en problemas de clasificación) o error cuadrático medio y r2 (en problemas de regresión).

- 5 Ajuste de hiperparámetros:

Se ajustan los hiperparámetros del modelo, como el número de árboles en el bosque, la profundidad máxima de cada árbol, etc., para mejorar el rendimiento del modelo en el conjunto de datos.


#### Arbol de decisón vs bosque aleatorio

La siguiente imagen muestra un modelo de arbol de decisión y bosque aleatorio para el mismo problema de regresión.
Ambos modelos tienen un error cuadrático medio (MSE) de entrenamiento  similar (bajo). 
Sin embargo, el error de generalización del random forest es mucho mejor debido al valor del mse en el conjunto de test (mas pequeño que el arbol de decisión). 
Esto se puede confirmar graficamente viendo que la predicción del modelo de random forest es más suave

Un argumento simple sobre por qué Bagging ayuda a bajar la pérdida por el error cuadrático, en resumen, es porque el promedio reduce la varianza y deja el sesgo sin cambios.

#### ventajas
* Mejor rendimiento que arboles de decisión
* Modelo no Parametrico (no hace supuestos sobre la distribución de los datos)
* Mas robusto ante outliers


#### Desventajas

* dificil interpretación
* costo computacional

### Boosting Models


El objetivo principal del algoritmo boosting es corregir los errores de predicción por medio de iteraciones en el entrenamiento de nuestro modelo. Los modelos son iterativamente ajustados y añadidos a nuestra arquitectura de tal manera que el modelo ajustado en la iteración  𝑖+1 esta enfocandose en las debilidades del modelo ajustado en la iteración  𝑖 y así sucesivamente.

El algoritmo de boosting suele ser implementado usando arboles de decisión podados, es decir, que solo se les permite tomar una cantidad limitada de decisiones (entre una y un par de hojas por arbol), a estos se les suele llamar weak learners o "stumps" en el contexto del aprendizaje estadistico, similar al metodo de bagging, las predicciones de estos modelos van a agregarse por medio de votación o promedios (aunque en este algoritmo suele tambien considerarse la precisión del modelo a la hora de determinar que tanto contribuye su voto, cosa que en bagging no suele hacerse). El objetivo de esto es crear un llamado strong learner por medio de muchos weak learners especializados.


A manera de resumen, debemos entender que boosting se refiere a una familia de algoritmos capaces de convertir un grupo de weak learners en un strong learner.

Los modelos más populares dentro de boosting son los siguientes:

#### AdaBoost (Adaptive Boosting):

- Introducido por Freund y Schapire en 1996.
- Se centra en corregir los errores de predicción de los modelos débiles anteriores dando más peso a las instancias mal clasificadas.
- Cada modelo débil se entrena de forma secuencial, y su peso en la predicción final depende de su rendimiento.


#### Gradient Boosting:

- Construye modelos en serie, donde cada nuevo modelo se enfoca en corregir los errores residuales del modelo anterior.
- Utiliza el gradiente descendente para minimizar una función de pérdida.

#### XGBoost (Extreme Gradient Boosting):

- Extensión del algoritmo Gradient Boosting que ha ganado popularidad por su eficiencia y rendimiento.
- Incorpora regularización L1 y L2 para controlar la complejidad del modelo.
- Soporta la paralelización y tiene implementaciones eficientes.

#### LightGBM:

- Desarrollado por Microsoft, es un algoritmo de boosting basado en gradientes.
- Utiliza la estrategia de crecimiento de árboles por hojas en lugar de por niveles, lo que mejora la eficiencia computacional.
- Es eficiente en el manejo de conjuntos de datos grandes.

#### CatBoost:

- Desarrollado por Yandex, está diseñado para manejar automáticamente variables categóricas sin necesidad de preprocesamiento adicional.
- Utiliza estrategias de regularización y técnicas para prevenir el sobreajuste.
- Tiene implementaciones eficientes y es fácil de usar.

### Ventajas:

- Mayor precisión: Boosting tiende a producir modelos más precisos en comparación con modelos individuales, ya que se enfoca en corregir los errores de los modelos anteriores.

- Manejo de datos desequilibrados: Boosting puede manejar eficazmente conjuntos de datos desequilibrados, donde una clase tiene menos instancias que otra.

- Reducción del overfitting: El boosting reduce el riesgo de sobreajuste (overfitting) al combinar múltiples modelos débiles, lo que mejora la generalización del modelo.

- Flexibilidad: Puede utilizarse con diversos tipos de modelos débiles, como árboles de decisión, lo que proporciona flexibilidad en la elección del modelo base.

- Selección automática de características: Algunos algoritmos de boosting pueden realizar selección automática de características, eliminando aquellas que no contribuyen significativamente al rendimiento del modelo.

### Desventajas:


- Tiempo de entrenamiento: El entrenamiento de modelos mediante boosting puede ser computacionalmente más costoso en términos de tiempo en comparación con algunos otros algoritmos más simples.

- Interpretabilidad: Los modelos de boosting pueden ser más difíciles de interpretar y explicar en comparación con modelos más simples, como regresiones lineales.

- Necesidad de ajuste de hiperparámetros: Los algoritmos de boosting tienen hiperparámetros que deben ser ajustados correctamente para obtener el mejor rendimiento, lo que puede requerir tiempo y experiencia.

- Riesgo de sobreajuste en conjuntos pequeños: En conjuntos de datos pequeños, existe el riesgo de sobreajuste, ya que el boosting puede aprender demasiado de las instancias individuales del conjunto de entrenamiento.

### Ejemplo Ensemble Learning Clasificación

```python
datalink="http://archive.ics.uci.edu/ml/machine-learning-databases/mammographic-masses/mammographic_masses.data"
df=pd.read_csv(StringIO(requests.get(datalink).text),na_values="?",header=None)
df.columns=["bi_rads","age","shape","margin","density","severity"]
df=df.drop(columns="bi_rads")
df["age"] = df["age"].fillna(df["age"].mean())
df["shape"] = df["shape"].fillna(4)
df["margin"] = df["margin"].fillna(1)
df["density"] = df["density"].fillna(3)
### convertimos a variables dummies
dummies_shape=pd.get_dummies(df["shape"],prefix="shape",drop_first=True).replace(True,1).replace(False,0)
dummies_margin=pd.get_dummies(df["margin"],prefix="margin",drop_first=True).replace(True,1).replace(False,0)
dummies_density=pd.get_dummies(df["density"],prefix="density",drop_first=True).replace(True,1).replace(False,0)
df = pd.concat([df,dummies_shape,dummies_margin,dummies_density],axis=1)
df=df.drop(columns=["shape","margin","density"])
X=df.drop(columns="severity")
Y=df["severity"]
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.3, random_state=3)
```

```python
tree_model = DecisionTreeClassifier().fit(X_train,Y_train)
forest_model = RandomForestClassifier(oob_score=True,random_state=0).fit(X_train,Y_train)
ada_model = AdaBoostClassifier().fit(X_train,Y_train)
xgb_model = XGBClassifier().fit(X_train,Y_train)

print("tree_model accuracy en train,",tree_model.score(X_train,Y_train))
print("tree_model accuracy en test,",tree_model.score(X_test,Y_test)) 
print("...........................")
print("forest_model accuracy en train,",forest_model.score(X_train,Y_train))
print("forest_model accuracy en test,",forest_model.score(X_test,Y_test)) 
print("...........................")
print("ada_model accuracy en train,",ada_model.score(X_train,Y_train))
print("ada_model accuracy en test,",ada_model.score(X_test,Y_test)) 
print("...........................")
print("xgb_model accuracy en train,",xgb_model.score(X_train,Y_train))
print("xgb_model accuracy en test,",xgb_model.score(X_test,Y_test)) 
```

el oob score(out of the bag) son registros que no se usan en la construcción de los arboles y se guardan para validar el modelo(como si fueran datos nuevos)

```python
forest_model.oob_score_
```

```python
model_params={
    "tree_model":{
        "model":DecisionTreeClassifier(), 
        "params":{

            "max_leaf_nodes":[4,7,8,10],
            "criterion":["gini","entropy"]
        }
    },
    "forest_model":{
        "model":RandomForestClassifier(), 
        "params":{
            "n_estimators":[10,100,500],
            "max_depth":[4,8,15],

            "criterion":["gini","entropy"],
            "random_state":[0,42]
        }
    },
    "ada_model":{
        "model":AdaBoostClassifier(), 
        "params":{
            "n_estimators":[10,100,500],
            "random_state":[0,42],
            "learning_rate":[.01,1,.001]
        }
    },
    "xgb_model":{
        "model":XGBClassifier(), 
        "params":{
            "n_estimators":[10,100,500],
            "learning_rate":[.01,1,.001]
        }
    },
    
            
}
```

```python
accuracy_models=[] # lista para guardar el accuracy de cada uno de los modelos
for model_name,mp in model_params.items():

    modelo=GridSearchCV(mp["model"],mp["params"],cv=5,n_jobs=-1,scoring="accuracy",verbose=0) #evaluamos con accuracy y partimos en 3 el conjunto de train
    
    modelo.fit(X_train,Y_train)
    accuracy_models.append({
        "model":model_name,
        "best_score":modelo.best_score_,
        "best_params":modelo.best_params_
    })
```

```python
### resumimos en una tabla los resultados obtenidos en gridsearch. se muestra el mejor modelo con un accuracy promedio de casi.80
df_accuracy_models=pd.DataFrame(accuracy_models).sort_values(by="best_score",ascending=False)
df_accuracy_models
```
