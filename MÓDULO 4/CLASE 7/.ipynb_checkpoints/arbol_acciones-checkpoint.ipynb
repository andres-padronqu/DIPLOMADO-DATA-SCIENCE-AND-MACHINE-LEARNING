El S&P 500, o Standard & Poor's 500, es un índice bursátil que mide el rendimiento de las 500 mayores empresas cotizadas en bolsa en los Estados Unidos. Este índice es considerado uno de los indicadores más representativos del mercado de valores estadounidense y, por extensión, del mercado financiero global.

Aquí hay algunas características clave del S&P 500:

Composición: El S&P 500 incluye empresas de diversos sectores, como tecnología, salud, finanzas, consumo discrecional, energía, entre otros. La selección de las empresas se basa en criterios específicos, como la capitalización bursátil, la liquidez y otros factores.

Ponderación por Capitalización: Las empresas dentro del S&P 500 no tienen igual peso. La ponderación se basa en la capitalización bursátil de cada empresa, lo que significa que las empresas más grandes tienen un impacto proporcionalmente mayor en el índice.

Representatividad del Mercado: Dada su diversidad y tamaño, el S&P 500 se considera un indicador representativo del mercado de valores estadounidense. Muchos inversores y gestores de fondos utilizan el S&P 500 como referencia para evaluar el rendimiento de sus carteras.

Cotización en Tiempo Real: La cotización del S&P 500 se actualiza en tiempo real durante las horas de negociación del mercado. La variación diaria del índice se utiliza para evaluar la salud general del mercado y como indicador de la confianza de los inversores.

Benchmark para Inversiones: El S&P 500 se utiliza como benchmark (punto de referencia) para evaluar el rendimiento de fondos de inversión, carteras de inversión y estrategias de gestión de activos. Los inversores a menudo comparan el rendimiento de sus inversiones con el S&P 500 para evaluar su éxito relativo.

Influencia en los Mercados Globales: Debido a la importancia económica de Estados Unidos y la representatividad del S&P 500, los movimientos en este índice pueden tener un impacto en los mercados financieros globales.

El S&P 500 es administrado por la empresa de servicios financieros Standard & Poor's, una división de S&P Global. Es uno de los índices más utilizados y seguidos en el mundo financiero, y su rendimiento a menudo se cita como una indicación del estado general del mercado de valores.

```python
df= pd.read_excel("muestra_acciones.xlsx")
df=df.set_index("timestamp")
```

```python
sp500_url="https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
sp500, *_=pd.read_html(sp500_url)
```

```python
sp500.head(5)
```

```python
companias=[]
for i in df:
  companias.append(i.replace("close_","").replace("volume_",""))
companias=list(set(companias))
companias
```

```python
sp500[sp500.Symbol.isin(companias)]
```

```python
df_close=df[[c for c in df.columns if "close" in c]]
df_close.tail()
```

```python
#el objetivo será calcular cúantos días lleva subiendo o cuántos días lleva bajando.
df_close=df_close.sort_index()
```

```python
close_count=df_close>df_close.shift(1) #día de mañana mayor que día de hoy
close_count.tail(5)
```

```python
close_count=close_count*1
close_count.tail(5)
```

```python
streaks=close_count != 0 #true son los días que subio
streaks.tail(5)
```

```python
streaks.cumsum().tail(5) #sumamos los días que subió
```

```python
streaks.cumsum().where(~streaks).tail(5)
#de la suma total préndeme los días en que cayó.


#aquí se identifican los días en que subió negándolo y que aparezca un na.
#una vez que identicamos un día que subió con "na" le pedimos que ponga el anterior.
```

```python
streaks.cumsum().where(~streaks).ffill().tail(5)
#se identifica la racha pasada. cn el bfill identificas en qué punto anterior rompió la racha
```

```python
streaks.cumsum()
```

```python
streaks.cumsum().where(~streaks).ffill()
```

```python
df_streaks=streaks.cumsum()-streaks.cumsum().where(~streaks).ffill()
#primer parte: streaks.cumsum() todos los días que subió
#segunda parte: streaks.cumsum().where(~streaks).ffill()
#identificar el día que subió y ponerle na y repetir el valor anterior

#de tal manera que cuando se reste si fue el primer día que subió cmo Verizon hoy, 11 de Julio lleva 2838 días subiendo
#en toda su historia (primera parte)
#segunda parte: identifica hoy que subió y pone NA y repite el anterior. Escribe 2837.

#la resta 2838-2837=1 día de racha.


df_streaks.tail(10)
```

```python
df_streaks=df_streaks.reset_index().melt(id_vars="timestamp")
```

```python
df_streaks ### cuantos dias llevaba la compañía de racha en esa fecha
```

```python
df_streaks["variable"]=df_streaks["variable"].apply(lambda v:v.replace("close_",""))
```

```python
df_streaks["GICS Sector"]=pd.merge(df_streaks,sp500,left_on="variable",right_on="Symbol",how="left")["GICS Sector"].tolist()
```

```python
df_streaks["y"]=(df_streaks["value"].shift(-1)>0)*1
#si un día después la acción subió pr+endeme de "1"
```

```python
df_streaks
```

```python
dummies_sector=pd.get_dummies(df_streaks["GICS Sector"],prefix="sector").replace(True,1).replace(False,0)
```

```python
df_streaks=pd.concat([df_streaks,dummies_sector],axis=1).drop(columns="GICS Sector")
```

```python
X=df_streaks[['value','sector_Communication Services',
       'sector_Consumer Discretionary', 'sector_Consumer Staples',
       'sector_Energy', 'sector_Financials', 'sector_Health Care',
       'sector_Industrials', 'sector_Information Technology',
       'sector_Materials', 'sector_Real Estate', 'sector_Utilities']]

Y= df_streaks[["y"]]
```

```python
X.head(5)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.3, random_state=3)
```

```python

tree_clf=DecisionTreeClassifier(random_state=9,min_samples_split=5,max_leaf_nodes=5)


tree_clf.fit(X_train,Y_train)
y_pred = tree_clf.predict(X_test)
print("accuracy en train,",tree_clf.score(X_train,Y_train))
print("accuracy en test,",tree_clf.score(X_test,Y_test))
```

```python
plot_tree(tree_clf.fit(X_train,Y_train),fontsize=8,filled=True,class_names=True,feature_names=X.columns)
```
