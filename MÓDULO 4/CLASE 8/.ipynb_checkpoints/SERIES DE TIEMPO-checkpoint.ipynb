Las series de tiempo son conjuntos de datos ordenados temporalmente, donde las observaciones se registran en intervalos regulares. Desde un punto de vista probabilistico , una serie de tiempo es la realización de un proceso estocástico.
La modelización de series temporales implica prever valores futuros basándose en patrones pasados. A diferencia de los problemas de clasificación y regresión que hemos ajustado anteriormente, en el analisis de series de tiempo nos encargaremos de hacer predicciones con base en una sola variable: El tiempo ($t$). Visto así queremos estimar un valor en función del tiempo $y=X(t)$


<img src="inflacion.png">

Una serie de tiempo se puede dividir en los siguientes componentes:

**Componente de tendencia**
Una tendencia es un movimiento direccional constante en una serie de tiempo. Estas tendencias serán deterministas o estocásticas. El primero nos permite proporcionar un fundamento para la tendencia, mientras que el segundo es una característica aleatoria de una serie que es poco probable que expliquemos. Las tendencias suelen aparecer en series financieras y muchos modelos comerciales utilizan sofisticados algoritmos de identificación de tendencias.

<img src="serie tendencia.png">


**Componente estacional**
Muchas series de tiempo contienen variaciones estacionales. Esto es particularmente cierto en las series que representan las ventas comerciales o los niveles climáticos. En las finanzas cuantitativas, a menudo vemos variaciones estacionales, particularmente en series relacionadas con las temporadas de vacaciones o la variación anual de la temperatura (como el gas natural). 

<img src="serie estacional.png">


Podemos escribir las componentes de una serie de tiempo yt como 

yt = St + Tt + Rt

donde St es el componente estacional, Tt es el componente de tendencia y Rt representa el componente restante de la serie de tiempo no capturado por el componente de tendencia o estacional.

## Definiciones clave en Series de tiempo

#### Ruido blanco

El ruido blanco es un concepto fundamental en estadísticas y series temporales. Se refiere a un tipo específico de secuencia de datos estocásticos que exhibe propiedades particulares. Aquí hay algunas características clave del ruido blanco:

- Media Constante:

La media del ruido blanco es constante y generalmente se establece en cero.

- Varianza Constante:

La varianza del ruido blanco es constante para todos los puntos de datos.

- Independencia:

Las observaciones de ruido blanco son independientes entre sí. La presencia o ausencia de un valor no proporciona información sobre los valores futuros o pasados.

-  Distribución Normal:

Las observaciones de ruido blanco se distribuyen de manera normal. Sin embargo, esta característica no es siempre estrictamente necesaria; en muchos casos, se asume que las observaciones son aproximadamente normales.


En resumen, el ruido blanco es un proceso estocástico que representa la aleatoriedad pura sin patrones o estructuras discernibles. A menudo, se utiliza como un modelo de base para describir el componente de error o residuo en muchos modelos estadísticos y de series temporales. Cuando los residuos de un modelo se comportan como ruido blanco, es un indicativo de que el modelo está capturando adecuadamente la variabilidad aleatoria en los datos.

<img src="ruido blanco.png">

##### Serie estacionaria

Una serie temporal se considera estacionaria cuando sus propiedades estadísticas no cambian con el tiempo. Si  la serie no tiene cambios sistemáticos en la media (no hay tendencia de la serie), si no tiene cambios sistematicos en la varianza (la varianza es constante a lo
largo del tiempo) y si no hay variaciones periodicas (no tiene ciclos) En otras palabras, la estacionariedad implica que la serie mantiene una estructura constante a lo largo de diferentes intervalos de tiempo. Así, una serie de tiempo con tendencia o con estacionalidad no es estacionaria, ya que la tendencia y la estacionalidad afectarán el valor de la serie de tiempo en diferentes momentos. Por otro lado, una serie de ruido blanco es estacionaria, ya que no importa cuando la observe; debería verse similar en cualquier momento.

Para muchos métodos de modelado de series temporales, como los modelos ARIMA (Autorregresivo Integrado de Medias Móviles), la estacionariedad es un requisito. Los modelos asumen que la serie es estacionaria o se puede convertir en estacionaria 

La prueba de Dickey-Fuller es comúnmente utilizada para evaluar la estacionariedad de una serie temporal. Si el valor p de la prueba es lo suficientemente bajo, se rechaza la hipótesis nula de no estacionariedad, indicando que la serie es estacionaria.



<img src="serie estacionaria.png">

#### Autocorrelacion

Hay muchas situaciones en las que los elementos consecutivos de una serie de tiempo muestran una correlación. Es decir, el comportamiento de los puntos secuenciales de la serie se afecta entre sí de manera dependiente. La autocorrelación es la similitud entre observaciones en función del desfase temporal entre ellas. 

#### Diferenciacion


La diferenciación es uno de los métodos utilizados para hacer estacionaria una serie de tiempo. En este método, calculamos la diferencia de términos consecutivos en la serie. 

La diferenciación se realiza típicamente para deshacerse de la media variable. Matemáticamente, la diferenciación se puede escribir como:

$y_t ′ = y_t - y_{\substack{t -1}}$

donde yt es el valor en un tiempo t.

La prueba de Dickey-Fuller es una prueba estadística utilizada para evaluar si una serie temporal no es estacionaria, lo que puede afectar la capacidad de modelar y predecir la serie de manera precisa 

La hipótesis nula de la prueba de Dickey-Fuller es que la serie temporal tiene una raíz unitaria, es decir, que no es estacionaria. La hipótesis alternativa es que la serie temporal es estacionaria.


$Ho:$ La serie no es estacionaria

$H1:$ La serie es estacionaria

El resultado de la prueba se compara con un valor crítico y se toma una decisión sobre si rechazar o no la hipótesis nula. Si el valor p (p-value) es lo suficientemente pequeño (generalmente menor que un umbral específico, como 0.05), se puede rechazar la hipótesis nula, lo que sugiere que la serie temporal es estacionaria.

```python
### hacemos la prueba de d-fuller aumentada para ver si la serie de tiempo es estacionaria

result = adfuller(df["SPY"])
print('ADF estadistico: %f' % result[0])
print('p-value: %f' % result[1])
if result[1]<.05:
    print("serie estacionaria")
else:
    print("serie no estacionaria")
    
#  es estacionaria 
```

```python
### hacemos la prueba de d-fuller con una diferencia 
result = adfuller(df["SPY"].diff().dropna())
print('ADF estadistico: %f' % result[0])
print('p-value: %f' % result[1])
if result[1]<.05:
    print("serie estacionaria")
else:
    print("serie no estacionaria")
    
#  es estacionaria 
```

### graficos de autocorrelacion y autocorrelacion parcial
La autocorrelación y la correlación parcial son herramientas gráficas útiles para analizar la dependencia entre las observaciones en una serie temporal. Ambos gráficos proporcionan información sobre la estructura de dependencia temporal en los datos.

```python
plot_acf(df["SPY"].diff(1).dropna(),lags=20)
```

```python
plot_pacf(df["SPY"].diff(1).dropna(),lags=20)
```

- Autocorrelación (ACF):

En el gráfico ACF, en el eje x (horizontal) se encuentran los diferentes lags (retrasos temporales), y en el eje y (vertical) se muestra el valor de autocorrelación en esos lags.

La autocorrelación mide la correlación entre una observación y sus observaciones anteriores a distintos retrasos. Un pico en el gráfico ACF en un cierto lag indica una correlación en ese punto.

- Correlación Parcial (PACF):

En el gráfico PACF, nuevamente en el eje x se encuentran los diferentes lags, y en el eje y se muestra el valor de correlación parcial.
La correlación parcial elimina la influencia de las observaciones intermedias, mostrando solo la relación directa entre una observación y otra con un cierto lag. Picos en el gráfico PACF indican una posible dependencia en ese lag.

En resumen, mientras que el gráfico ACF muestra la correlación total en un lag, el gráfico PACF muestra solo la correlación que no se puede explicar por lags intermedios. 

Estos gráficos son herramientas útiles para identificar posibles componentes AR (autorregresivos) y MA (media móvil) en un modelo ARIMA.

#### SUAVIZAMIENTO EXPONENCIAL SIMPLE

```python
# ajuste de parametros
list_smoothing=[]
list_rmse = []
list_aic = []
list_mae = []
modelos_exp =pd.DataFrame()
for i in np.arange(0.1,1,.1):

    mod_exp = SimpleExpSmoothing(train).fit(smoothing_level=round(i,1), optimized=False)
    fc = mod_exp.forecast(len(test))
    rmse = np.sqrt(mean_squared_error(test, fc.tolist()))
    mae = mean_absolute_error(test, fc.tolist())

    list_aic.append(mod_exp.aic)
    list_smoothing.append(i)
    list_rmse.append(rmse)
    list_mae.append(mae)
modelos_exp["smoothing_level"]=list_smoothing
modelos_exp["rmse"]=list_rmse
modelos_exp["aic"]=list_aic
modelos_exp["mae"]=list_mae

modelos_exp.sort_values(by="aic")
```

```python
model1 = SimpleExpSmoothing(train).fit(smoothing_level=0.9, optimized=False)
fc1 = model1.forecast(len(test)).rename("alpha=0.5")

model2 = SimpleExpSmoothing(train).fit(smoothing_level=0.8, optimized=False)
fc2 = model2.forecast(len(test)).rename("alpha=0.6")

model3 = SimpleExpSmoothing(train).fit(smoothing_level=0.7, optimized=False)
fc3 = model3.forecast(len(test)).rename("alpha=0.7")

plt.figure(figsize=(12, 8))
plt.plot(train, marker="o", color="black")
plt.plot(model1.fittedvalues, marker="o", color="blue")
(line1,) = plt.plot(fc1, marker="o", color="blue")
plt.plot(model2.fittedvalues, marker="o", color="red")
(line2,) = plt.plot(fc2, marker="o", color="red")
plt.plot(model3.fittedvalues, marker="o", color="green")
(line3,) = plt.plot(fc3, marker="o", color="green")
plt.legend([line1, line2,line3], [fc1.name, fc2.name,fc3.name])
```

### HOLT WINTERS

```python
# ajuste de parametros
list_smoothing_level=[]
list_smoothing_slope=[]
list_rmse = []
list_aic = []
list_mae = []
modelos_holt =pd.DataFrame()
for i in np.arange(0.1,1,.1):
    for j in np.arange(0.1,1,.1):

        mod_holt = Holt(train).fit(smoothing_level=round(i,1),smoothing_slope=round(j,1))
        fc = mod_holt.forecast(len(test))
        rmse = np.sqrt(mean_squared_error(test, fc.tolist()))
        mae= mean_absolute_error(test, fc.tolist())
        


        list_smoothing_level.append(i)
        list_smoothing_slope.append(j)
        list_aic.append(mod_holt.aic)
        list_rmse.append(rmse)
        list_mae.append(mae)
modelos_holt["smoothing_level"]=list_smoothing_level
modelos_holt["smoothing_slope"]=list_smoothing_slope
modelos_holt["rmse"]=list_rmse
modelos_holt["aic"]=list_aic
modelos_holt["mae"]=list_mae
modelos_holt.sort_values(by="aic")
```

```python
model1 = Holt(train).fit(smoothing_level=0.9,smoothing_slope=.1)
fc1 = model1.forecast(len(test)).rename("smoothing_level = .9,smoothing_slope=.1")
model2 = Holt(train).fit(smoothing_level=0.8,smoothing_slope=.1)
fc2 = model2.forecast(len(test)).rename("smoothing_level = .8,smoothing_slope=.1")
model3 = Holt(train).fit(smoothing_level=0.9,smoothing_slope=.2)
fc3 = model3.forecast(len(test)).rename("smoothing_level = .9,smoothing_slope=.2")


plt.figure(figsize=(12, 8))
plt.plot(train, marker="o", color="black")
plt.plot(model1.fittedvalues, color="blue")
(line1,) = plt.plot(fc1, marker="o", color="blue")
plt.plot(model2.fittedvalues, color="red")
(line2,) = plt.plot(fc2, marker="o", color="red")
plt.plot(model3.fittedvalues, color="green")
(line3,) = plt.plot(fc3, marker="o", color="green")

plt.legend([line1, line2,line3], [fc1.name, fc2.name,fc3.name])
```

#### suavizamiento exponencial estacional

```python
model1 = ExponentialSmoothing(train,trend="add").fit()
fc1 = model1.forecast(len(test)).rename("trend")

model2 = ExponentialSmoothing(train,trend="add",seasonal="add",seasonal_periods=12).fit()
fc2 = model2.forecast(len(test)).rename("trend,seasonal")



plt.figure(figsize=(12, 8))
plt.plot(train, marker="o", color="black")
plt.plot(model1.fittedvalues, marker="o", color="blue")
(line1,) = plt.plot(fc1, marker="o", color="blue")
plt.plot(model2.fittedvalues, marker="o", color="red")
(line2,) = plt.plot(fc2, marker="o", color="red")

plt.legend([line1, line2], [fc1.name, fc2.name])
```
